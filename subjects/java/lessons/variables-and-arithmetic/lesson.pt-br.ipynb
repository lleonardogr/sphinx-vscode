## 1. Um programa roda de cima para baixo

## 2. Variáveis: caixas com nome para valores

Uma **variável** é uma caixa com nome que guarda um valor. Você a cria uma vez, com o **tipo**, o nome e o primeiro valor:

```java
int eggs = 12;
```

Depois disso usa só o nome: para ler o valor (`eggs + 1`) ou para pôr um valor novo na caixa (`eggs = 15;`). O valor antigo é substituído. Os slides mostram onde cada variável mora na memória: números ficam direto na **stack**, enquanto o texto de uma `String` mora no **heap** e a variável guarda uma seta para ele.

#### 🎞️ Variáveis na stack e no heap

![Cada variável ganha uma caixa na stack: números ficam guardados ali mesmo, enquanto uma String mora no heap e a variável guarda uma referência a ela; copiar uma referência compartilha o objeto](variables-memory-steps.pt-br.svg)

````java
int eggs = 12;
IO.println(eggs);
eggs = 15;           // um valor novo substitui o antigo
IO.println(eggs);
eggs = eggs + 1;     // lê o valor antigo, soma 1, guarda o resultado
IO.println(eggs);

int copy = eggs;     // copia o valor: duas caixas separadas
copy = 100;
IO.println(eggs + " e " + copy);
````

## 3. Tipos: o que cabe em cada caixa

Toda variável tem um **tipo**, e só valores desse tipo cabem nela:

| Tipo | Guarda | Exemplos |
|---|---|---|
| `int` | números inteiros | `12`, `-3`, `0` |
| `double` | números com casas decimais | `2.5`, `-0.1`, `3.0` |
| `String` | texto | `"Ada"`, `""` |
| `boolean` | verdadeiro ou falso | `true`, `false` |
| `char` | um caractere, entre aspas simples | `'A'`, `'7'` |

Um `int` cabe num `double` (ganha `.0`), mas um `double` não cabe num `int`: o Java não joga as casas decimais fora a não ser que você peça com `(int)`.

````java
int people = 4;
double price = 7.5;
String city = "Recife";
boolean open = true;
char grade = 'A';
IO.println(people + " pessoas, " + price + " cada, em " + city + ", aberto: " + open + ", nota " + grade);

double asDecimal = people;     // um int cabe num double
int cut = (int) 9.99;          // (int) descarta as casas decimais; não arredonda
IO.println(asDecimal + " " + cut);
````

## 4. Aritmética: + - * / e a ordem das operações

O Java faz contas como uma calculadora: `+`, `-`, `*` (vezes) e `/` (dividido por). Como na matemática, `*` e `/` vêm **antes** de `+` e `-`, e os parênteses mudam a ordem: `2 + 3 * 4` é `14`, enquanto `(2 + 3) * 4` é `20`.

````java
IO.println(2 + 3 * 4);
IO.println((2 + 3) * 4);
IO.println(10 - 4 - 3);    // da esquerda para a direita: (10 - 4) - 3

double width = 2.5;
double height = 4;
IO.println("Área: " + width * height);
````

## 5. Divisão inteira e resto

Aqui está a surpresa: quando **os dois** lados de `/` são `int`, o resultado também é `int`, e as casas decimais são **descartadas**. `17 / 6` é `2`, não `2.83`. A outra metade dessa divisão é o `%`, o **resto**: `17 % 6` é `5`, porque 17 = 6 × 2 + 5.

Para ter as casas decimais, faça um dos lados ser `double`: `17 / 6.0`, ou `(double) a / b` com variáveis. O `%` também é útil sozinho: `n % 2 == 0` diz se `n` é par, e `n % 10` é o último dígito.

````java
int eggs = 17;
IO.println("Caixas de 6: " + eggs / 6);
IO.println("Sobram:      " + eggs % 6);
IO.println("Exato:       " + eggs / 6.0);

int a = 7;
int b = 2;
IO.println(a / b);              // int / int
IO.println((double) a / b);     // double / int
IO.println(1234 % 10);          // o último dígito
````

## 6. Juntando texto com +

O `+` tem dois trabalhos. Entre números ele soma; assim que um dos lados é uma `String`, ele **junta** texto. O Java lê da esquerda para a direita, então `1 + 2 + "3"` soma primeiro (`"33"`), enquanto `"1" + 2 + 3` junta tudo (`"123"`). Use parênteses para somar dentro de uma mensagem: `"Total: " + (a + b)`.

````java
String name = "Ada";
IO.println("Oi, " + name + "!");
IO.println(1 + 2 + "3");
IO.println("1" + 2 + 3);

int a = 4;
int b = 5;
IO.println("Total: " + a + b);       // junta 4 e depois 5
IO.println("Total: " + (a + b));     // soma antes
````

## 7. De texto digitado para números

Nos desafios, `IO.readln()` lê uma linha que o usuário digitou. Ele sempre devolve uma **String**, mesmo que o usuário tenha digitado algarismos: `"42"` é texto, não número. Converta com `Integer.parseInt(texto)` para um `int` ou `Double.parseDouble(texto)` para um `double`. Se o texto não for um número, o programa para com uma `NumberFormatException`.

As células do notebook não leem o teclado, então abaixo o texto "digitado" está escrito direto.

````java
String typed = "42";                 // o que o IO.readln() devolveria
IO.println(typed + 1);                // texto: junta
int number = Integer.parseInt(typed);
IO.println(number + 1);               // número: soma

double height = Double.parseDouble("1.75");
IO.println(height * 2);
````

## 8. Atalhos: +=, ++ e --

Mudar uma variável a partir do próprio valor é tão comum que o Java tem atalhos: `total += 5` significa `total = total + 5` (também `-=`, `*=`, `/=`), e `count++` soma 1 (`count--` subtrai 1). Você vai usar isso o tempo todo em laços.

````java
int total = 10;
total += 5;
IO.println(total);
total *= 2;
IO.println(total);
int count = 0;
count++;
count++;
count--;
IO.println(count);
````

## Em resumo

Um programa roda suas linhas de cima para baixo. Uma **variável** é uma caixa com nome para um valor, e o **tipo** dela diz o que cabe ali: `int` para números inteiros, `double` para números com decimais, `String` para texto.

```java
void main() {
    String name = IO.readln();
    int eggs = Integer.parseInt(IO.readln());
    IO.println("Oi, " + name + "!");
    IO.println("Caixas de 6: " + eggs / 6);
    IO.println("Sobram: " + eggs % 6);
    IO.println("Meia caixa: " + 6 / 2.0);
}
```

- `+` soma números e junta textos.
- `/` entre dois `int`s fica só com a parte inteira: `17 / 6` é `2`. `%` dá o resto: `17 % 6` é `5`.
- `IO.readln()` lê uma linha como texto; `Integer.parseInt` e `Double.parseDouble` a transformam em número.

**Cuidado:** `7 / 2` é `3`, não `3.5`. Escreva `7 / 2.0` quando precisar dos decimais.
