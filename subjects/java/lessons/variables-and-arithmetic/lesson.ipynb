## 1. A program runs from top to bottom

A program is a list of instructions, and Java runs them **in order, from top to bottom**, one line at a time. `IO.println(...)` prints what is in the parentheses and moves to the next line; `IO.print(...)` prints and stays on the same line. Text goes between double quotes: `"like this"`. Every instruction ends with `;`.

````java
IO.println("First line");
IO.println("Second line");
IO.print("Same ");
IO.print("line");
IO.println();   // just ends the line
````

## 2. Variables: named boxes for values

A **variable** is a box with a name that holds a value. You create it once, with its **type**, its name and its first value:

```java
int eggs = 12;
```

After that you use just the name: to read the value (`eggs + 1`) or to put a new one in the box (`eggs = 15;`). The old value is replaced. The slides show where each variable lives in memory: numbers sit right in the **stack**, while a `String`'s text lives in the **heap** and the variable holds an arrow to it.

#### 🎞️ Variables in the stack and the heap

![Each variable gets a box in the stack: numbers are stored right there, while a String lives in the heap and the variable holds a reference to it; copying a reference shares the object](variables-memory-steps.svg)

````java
int eggs = 12;
IO.println(eggs);
eggs = 15;           // a new value replaces the old one
IO.println(eggs);
eggs = eggs + 1;     // read the old value, add 1, store the result
IO.println(eggs);

int copy = eggs;     // copies the value: two separate boxes
copy = 100;
IO.println(eggs + " and " + copy);
````

## 3. Types: what fits in each box

Every variable has a **type**, and only values of that type fit in it:

| Type | Holds | Examples |
|---|---|---|
| `int` | whole numbers | `12`, `-3`, `0` |
| `double` | numbers with decimals | `2.5`, `-0.1`, `3.0` |
| `String` | text | `"Ada"`, `""` |
| `boolean` | true or false | `true`, `false` |
| `char` | one character, in single quotes | `'A'`, `'7'` |

An `int` fits in a `double` (it gets `.0`), but a `double` doesn't fit in an `int`: Java won't throw the decimals away unless you ask with `(int)`.

````java
int people = 4;
double price = 7.5;
String city = "Recife";
boolean open = true;
char grade = 'A';
IO.println(people + " people, " + price + " each, in " + city + ", open: " + open + ", grade " + grade);

double asDecimal = people;     // an int fits in a double
int cut = (int) 9.99;          // (int) drops the decimals; it doesn't round
IO.println(asDecimal + " " + cut);
````

## 4. Arithmetic: + - * / and the order of operations

Java does arithmetic like a calculator: `+`, `-`, `*` (times) and `/` (divided by). As in math, `*` and `/` come **before** `+` and `-`, and parentheses change the order: `2 + 3 * 4` is `14`, while `(2 + 3) * 4` is `20`.

````java
IO.println(2 + 3 * 4);
IO.println((2 + 3) * 4);
IO.println(10 - 4 - 3);    // left to right: (10 - 4) - 3

double width = 2.5;
double height = 4;
IO.println("Area: " + width * height);
````

## 5. Whole-number division and the remainder

Here is the surprise: when **both** sides of `/` are `int`s, the result is an `int` too, and the decimals are **dropped**. `17 / 6` is `2`, not `2.83`. The other half of that division is `%`, the **remainder**: `17 % 6` is `5`, because 17 = 6 × 2 + 5.

To get the decimals, make one side a `double`: `17 / 6.0`, or `(double) a / b` with variables. `%` is also handy on its own: `n % 2 == 0` tells whether `n` is even, and `n % 10` is its last digit.

````java
int eggs = 17;
IO.println("Boxes of 6: " + eggs / 6);
IO.println("Left over:  " + eggs % 6);
IO.println("Exact:      " + eggs / 6.0);

int a = 7;
int b = 2;
IO.println(a / b);              // int / int
IO.println((double) a / b);     // double / int
IO.println(1234 % 10);          // the last digit
````

## 6. Joining text with +

`+` has two jobs. Between numbers it adds; as soon as one side is a `String`, it **joins** text. Java reads from left to right, so `1 + 2 + "3"` adds first (`"33"`), while `"1" + 2 + 3` joins all the way (`"123"`). Use parentheses to add inside a message: `"Total: " + (a + b)`.

````java
String name = "Ada";
IO.println("Hi, " + name + "!");
IO.println(1 + 2 + "3");
IO.println("1" + 2 + 3);

int a = 4;
int b = 5;
IO.println("Total: " + a + b);       // joins 4 and then 5
IO.println("Total: " + (a + b));     // adds first
````

## 7. From typed text to numbers

In the challenges, `IO.readln()` reads a line the user typed. It always gives a **String**, even if the user typed digits: `"42"` is text, not a number. Convert it with `Integer.parseInt(text)` for an `int` or `Double.parseDouble(text)` for a `double`. If the text isn't a number, the program stops with a `NumberFormatException`.

Notebook cells can't read the keyboard, so below the "typed" text is written directly.

````java
String typed = "42";                 // what IO.readln() would give
IO.println(typed + 1);                // text: joins
int number = Integer.parseInt(typed);
IO.println(number + 1);               // number: adds

double height = Double.parseDouble("1.75");
IO.println(height * 2);
````

🧪 **Try it:** change `"42"` to `"4x2"` and run: read the exception's message. Then try `" 42"` with a space, and fix it with `typed.trim()`.

## 8. Shortcuts: +=, ++ and --

Changing a variable based on its own value is so common that Java has shortcuts: `total += 5` means `total = total + 5` (also `-=`, `*=`, `/=`), and `count++` adds 1 (`count--` subtracts 1). You'll use them all the time in loops.

````java
int total = 10;
total += 5;
IO.println(total);
total *= 2;
IO.println(total);
int count = 0;
count++;
count++;
count--;
IO.println(count);
````

## In short

A program runs its lines from top to bottom. A **variable** is a named box for a value, and its **type** says what fits in it: `int` for whole numbers, `double` for numbers with decimals, `String` for text.

```java
void main() {
    String name = IO.readln();
    int eggs = Integer.parseInt(IO.readln());
    IO.println("Hi, " + name + "!");
    IO.println("Boxes of 6: " + eggs / 6);
    IO.println("Left over: " + eggs % 6);
    IO.println("Half a box: " + 6 / 2.0);
}
```

- `+` adds numbers and joins text.
- `/` between two `int`s keeps only the whole part: `17 / 6` is `2`. `%` gives the remainder: `17 % 6` is `5`.
- `IO.readln()` reads a line as text; `Integer.parseInt` and `Double.parseDouble` turn it into a number.

**Watch out:** `7 / 2` is `3`, not `3.5`. Write `7 / 2.0` when you need the decimals.
