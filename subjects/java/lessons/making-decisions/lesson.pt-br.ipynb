## Em resumo

`if` roda um bloco só quando a condição dele é `true`. As condições comparam valores com `==`, `!=`, `<`, `<=`, `>` e `>=`, e se combinam com `&&` (e), `||` (ou) e `!` (não).

```java
void main() {
    int age = Integer.parseInt(IO.readln());
    int day = Integer.parseInt(IO.readln()); // 1 = segunda … 7 = domingo
    if (age < 12) {
        IO.println("Ingresso infantil");
    } else if (age >= 65 || day == 3) {
        IO.println("Ingresso com desconto");
    } else {
        IO.println("Ingresso inteiro");
    }
    String kind = switch (day) {
        case 6, 7 -> "fim de semana";
        default -> "dia de semana";
    };
    IO.println(age >= 18 ? "Adulto num " + kind : "Menor num " + kind);
}
```

O Java testa o `if` e depois cada `else if`, de cima para baixo, e roda só o **primeiro** bloco cuja condição é verdadeira.

#### 🎞️ Uma cadeia if / else if, passo a passo

![Com age 70, o Java testa age < 12 (falso), depois age >= 65 (verdadeiro), roda esse bloco e pula o else](if-chain-steps.pt-br.svg)

**Experimente outros valores:** mude `age` e `day` e rode de novo. `&&` para no primeiro `false`, então a divisão abaixo nunca roda quando `count` é 0.

````java
int age = 70;
int day = 3; // 1 = segunda ... 7 = domingo
String ticket;
if (age < 12) {
    ticket = "infantil";
} else if (age >= 65) {
    ticket = "idoso";
} else {
    ticket = "inteiro";
}
String when = switch (day) {
    case 6, 7 -> "fim de semana";
    default -> "dia útil";
};
IO.println("ingresso " + ticket + " em " + when);

int count = 0;
if (count != 0 && 100 / count > 5) {
    IO.println("nunca impresso");
}
IO.println("sem divisão por zero");
````
