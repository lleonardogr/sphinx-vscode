## 1. Verdadeiro ou falso: comparando valores

Um programa toma decisões **comparando valores**. Toda comparação dá um `boolean`, um valor que só pode ser `true` (verdadeiro) ou `false` (falso):

| Operador | Significa | Exemplo com `age = 15` |
|---|---|---|
| `==` | igual a | `age == 15` → `true` |
| `!=` | diferente de | `age != 15` → `false` |
| `<` / `>` | menor que / maior que | `age > 18` → `false` |
| `<=` / `>=` | menor ou igual / maior ou igual | `age <= 15` → `true` |

Repare na diferença: um `=` sozinho **guarda** um valor numa variável (`age = 15`), enquanto `==` **pergunta** se dois valores são iguais.

````java
int age = 15;
IO.println(age >= 18);   // 15 é pelo menos 18?
IO.println(age < 18);
IO.println(age == 15);
IO.println(age != 15);

boolean adult = age >= 18;   // o resultado de uma comparação pode ser guardado, como qualquer valor
IO.println("adulto? " + adult);
````

## 2. if: rodar um bloco só às vezes

`if` (se) roda um bloco de código **só quando a condição dele é `true`**:

```java
if (condição) {
    // roda só quando a condição é true
}
// o programa sempre continua aqui
```

A condição vai entre parênteses e precisa ser um `boolean`. O bloco vai entre chaves `{ }`. Quando a condição é `false`, o Java pula o bloco inteiro e segue para a linha depois dele.

````java
int temperature = 32;
IO.println("Olhando o tempo...");
if (temperature > 30) {
    IO.println("Está quente: beba água!");
}
IO.println("Pronto.");   // fora do if: sempre roda
````

🧪 **Experimente:** mude `temperature` para `25`. Qual linha some, e quais continuam aparecendo? O que acontece com exatamente `30`?

**Use sempre as chaves**, mesmo para uma linha só. Sem elas, só a primeira linha depois do `if` pertence a ele, e uma segunda linha acrescentada depois rodaria sempre.

## 3. if e else: um caminho ou o outro

`else` (senão) dá o caminho para quando a condição é `false`. **Exatamente um** dos dois blocos roda: nunca os dois, nunca nenhum.

```java
if (condição) {
    // quando é true
} else {
    // quando é false
}
```

````java
int grade = 6;
if (grade >= 7) {
    IO.println("Aprovado!");
} else {
    IO.println("Recuperação");
}
````

## 4. else if: escolhendo entre vários caminhos

Com mais de dois casos, encadeie as condições com `else if` (senão, se). O Java as testa **de cima para baixo** e roda só o **primeiro** bloco cuja condição é `true`. O resto é pulado, mesmo que também fosse verdadeiro. O `else` final é opcional: ele pega tudo o que não combinou com nada acima.

#### 🎞️ Uma cadeia if / else if, passo a passo

![Com age 70, o Java testa age < 12 (falso), depois age >= 65 (verdadeiro), roda esse bloco e pula o else](if-chain-steps.pt-br.svg)

````java
int age = 70;
String ticket;
if (age < 12) {
    ticket = "infantil";
} else if (age >= 65) {
    ticket = "idoso";
} else {
    ticket = "inteira";
}
IO.println("Ingresso: " + ticket);
````

🧪 **Experimente:** rode com `5`, `11`, `12`, `40`, `65` e `70`, prevendo cada resposta antes. Qual condição decide cada uma?

**A ordem importa.** Aqui o teste geral vem primeiro, então o específico nunca é alcançado:

````java
int age = 70;
if (age >= 18) {
    IO.println("adulto");
} else if (age >= 65) {
    IO.println("idoso");   // nunca roda: 70 já combinou com age >= 18
}
````

## 5. Combinando condições: &&, || e !

As condições podem ser combinadas:

| Operador | Nome | É `true` quando… |
|---|---|---|
| `a && b` | e | **as duas** são verdadeiras |
| `a \|\| b` | ou | **pelo menos uma** é verdadeira |
| `!a` | não | `a` é falsa (ele inverte o valor) |

O `&&` para no primeiro `false`, e o `||` para no primeiro `true`: o resto nem é avaliado. Isso se chama **curto-circuito**, e protege testes que falhariam, como dividir por zero.

````java
int age = 20;
boolean student = true;
IO.println(age >= 18 && student);   // as duas são verdadeiras
IO.println(age < 12 || student);    // uma basta
IO.println(!student);               // invertido

int count = 0;
int total = 50;
if (count != 0 && total / count > 5) {
    IO.println("média acima de 5");
} else {
    IO.println("sem dados, ou média pequena");
}
````

## 6. Comparando textos: equals, não ==

Para `String`s, o `==` pergunta se as duas variáveis apontam para **o mesmo objeto** na memória, não se têm as mesmas letras. Dois textos com letras iguais podem ser objetos diferentes, como acontece com um texto digitado pelo usuário. Para comparar as letras, use `equals`, ou `equalsIgnoreCase` para ignorar maiúsculas e minúsculas.

````java
String answer = "sim";
String typed = new String("sim");   // um objeto separado, como um texto lido do teclado
IO.println(typed == answer);        // mesmo objeto?
IO.println(typed.equals(answer));   // mesmas letras?
IO.println("SIM".equalsIgnoreCase(answer));

if (typed.equals("sim")) {
    IO.println("Confirmado");
}
````

## 7. switch: escolhendo pelo valor

Quando você compara **um valor** com uma lista de opções fixas, o `switch` fica mais claro que uma longa cadeia de `else if`. Cada `case` lista um ou mais valores, `->` dá o resultado, e `default` cobre todo o resto:

```java
String kind = switch (day) {
    case 6, 7 -> "fim de semana";
    default -> "dia útil";
};
```

Um `switch` assim é uma **expressão**: ele produz um valor, então precisa cobrir toda entrada possível (é para isso que serve o `default`) e termina com `;`.

````java
int day = 6;   // 1 = segunda ... 7 = domingo
String kind = switch (day) {
    case 1, 2, 3, 4, 5 -> "dia útil";
    case 6, 7 -> "fim de semana";
    default -> "dia inválido";
};
IO.println(day + " é " + kind);

String command = "parar";
switch (command) {   // um switch também pode só rodar código, e funciona com Strings
    case "andar" -> IO.println("Andando");
    case "parar" -> IO.println("Parado");
    default -> IO.println("Comando desconhecido");
}
````

## 8. O operador ternário: um if / else curto

`condição ? valorSeVerdadeiro : valorSeFalso` escolhe um de dois **valores**. É um `if` / `else` curto que cabe dentro de uma expressão, bom para escolhas pequenas como singular ou plural. Para algo maior, um `if` normal fica mais legível.

````java
int items = 1;
String word = items == 1 ? "item" : "itens";
IO.println(items + " " + word);

int score = 72;
IO.println("Resultado: " + (score >= 60 ? "aprovado" : "reprovado"));
````

## 9. Erros comuns

- **Um `;` logo depois de `if (…)`** encerra o `if` ali: o bloco de baixo vira um bloco comum, que sempre roda. Rode a célula abaixo e veja.
- **`=` no lugar de `==`** numa condição: `if (age = 18)` nem compila, porque `age = 18` é um `int`, não um `boolean`.
- **`==` com Strings:** use `equals` (passo 6).
- **Condições na ordem errada:** o caso específico primeiro (passo 4).

````java
int x = 5;
if (x > 10); {
    IO.println("x é grande?");   // imprime mesmo assim: o ; encerrou o if
}
````

🧪 **Experimente:** tire o `;` depois de `if (x > 10)` e rode de novo.

## Em resumo

`if` roda um bloco só quando a condição dele é `true`. As condições comparam valores com `==`, `!=`, `<`, `<=`, `>` e `>=`, e se combinam com `&&` (e), `||` (ou) e `!` (não).

```java
void main() {
    int age = Integer.parseInt(IO.readln());
    int day = Integer.parseInt(IO.readln()); // 1 = segunda … 7 = domingo
    if (age < 12) {
        IO.println("Ingresso infantil");
    } else if (age >= 65 || day == 3) {
        IO.println("Ingresso com desconto");
    } else {
        IO.println("Ingresso inteiro");
    }
    String kind = switch (day) {
        case 6, 7 -> "fim de semana";
        default -> "dia de semana";
    };
    IO.println(age >= 18 ? "Adulto num " + kind : "Menor num " + kind);
}
```

O Java testa o `if` e depois cada `else if`, de cima para baixo, e roda só o **primeiro** bloco cuja condição é verdadeira.

**Cuidado:** a ordem importa. Com `if (age >= 18)` antes de `if (age >= 65)`, o segundo teste nunca é alcançado.
