## In short

`if` runs a block only when its condition is `true`. Conditions compare values with `==`, `!=`, `<`, `<=`, `>` and `>=`, and combine them with `&&` (and), `||` (or) and `!` (not).

```java
void main() {
    int age = Integer.parseInt(IO.readln());
    int day = Integer.parseInt(IO.readln()); // 1 = Monday … 7 = Sunday
    if (age < 12) {
        IO.println("Child ticket");
    } else if (age >= 65 || day == 3) {
        IO.println("Discount ticket");
    } else {
        IO.println("Full ticket");
    }
    String kind = switch (day) {
        case 6, 7 -> "weekend";
        default -> "weekday";
    };
    IO.println(age >= 18 ? "Adult on a " + kind : "Minor on a " + kind);
}
```

Java tests `if`, then each `else if`, from top to bottom, and runs only the **first** block whose condition is true.

#### 🎞️ An if / else if chain, step by step

![With age 70, Java tests age < 12 (false), then age >= 65 (true), runs that block and skips the else](if-chain-steps.svg)

**Try different values:** change `age` and `day` and run again. `&&` stops at the first `false`, so the division below never runs when `count` is 0.

````java
int age = 70;
int day = 3; // 1 = Monday ... 7 = Sunday
String ticket;
if (age < 12) {
    ticket = "child";
} else if (age >= 65) {
    ticket = "senior";
} else {
    ticket = "full";
}
String when = switch (day) {
    case 6, 7 -> "weekend";
    default -> "weekday";
};
IO.println(ticket + " ticket on a " + when);

int count = 0;
if (count != 0 && 100 / count > 5) {
    IO.println("never printed");
}
IO.println("no division by zero");
````
