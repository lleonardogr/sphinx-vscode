## 1. true or false: comparing values

A program makes decisions by **comparing values**. Every comparison gives a `boolean`, a value that is either `true` or `false`:

| Operator | Means | Example with `age = 15` |
|---|---|---|
| `==` | equal to | `age == 15` → `true` |
| `!=` | not equal to | `age != 15` → `false` |
| `<` / `>` | less than / greater than | `age > 18` → `false` |
| `<=` / `>=` | less or equal / greater or equal | `age <= 15` → `true` |

Watch the difference: a single `=` **stores** a value in a variable (`age = 15`), while `==` **asks** whether two values are equal.

````java
int age = 15;
IO.println(age >= 18);   // is 15 at least 18?
IO.println(age < 18);
IO.println(age == 15);
IO.println(age != 15);

boolean adult = age >= 18;   // a comparison's result can be stored, like any value
IO.println("adult? " + adult);
````

## 2. if: running a block only sometimes

`if` runs a block of code **only when its condition is `true`**:

```java
if (condition) {
    // runs only when condition is true
}
// the program always continues here
```

The condition goes in parentheses and must be a `boolean`. The block goes between braces `{ }`. When the condition is `false`, Java skips the whole block and goes on with the line after it.

````java
int temperature = 32;
IO.println("Checking the weather...");
if (temperature > 30) {
    IO.println("It's hot: drink some water!");
}
IO.println("Done.");   // outside the if: always runs
````

🧪 **Try it:** change `temperature` to `25`. Which line disappears, and which ones are still printed? What happens with exactly `30`?

**Always use braces**, even for a single line. Without them, only the first line after the `if` belongs to it, and a second line added later would run every time.

## 3. if and else: one path or the other

`else` (otherwise) gives the path for when the condition is `false`. **Exactly one** of the two blocks runs: never both, never neither.

```java
if (condition) {
    // when it's true
} else {
    // when it's false
}
```

````java
int grade = 6;
if (grade >= 7) {
    IO.println("Passed!");
} else {
    IO.println("Make-up exam");
}
````

## 4. else if: choosing among several paths

With more than two cases, chain the conditions with `else if`. Java tests them **from top to bottom** and runs only the **first** block whose condition is `true`. The rest are skipped, even if they would also be true. The final `else` is optional: it catches everything that matched nothing above.

#### 🎞️ An if / else if chain, step by step

![With age 70, Java tests age < 12 (false), then age >= 65 (true), runs that block and skips the else](if-chain-steps.svg)

````java
int age = 70;
String ticket;
if (age < 12) {
    ticket = "child";
} else if (age >= 65) {
    ticket = "senior";
} else {
    ticket = "full price";
}
IO.println("Ticket: " + ticket);
````

🧪 **Try it:** run it with `5`, `11`, `12`, `40`, `65` and `70`, predicting each answer first. Which condition decides each one?

**Order matters.** Here the general test comes first, so the specific one is never reached:

````java
int age = 70;
if (age >= 18) {
    IO.println("adult");
} else if (age >= 65) {
    IO.println("senior");   // never runs: 70 already matched age >= 18
}
````

## 5. Combining conditions: &&, || and !

Conditions can be combined:

| Operator | Name | It's `true` when… |
|---|---|---|
| `a && b` | and | **both** are true |
| `a \|\| b` | or | **at least one** is true |
| `!a` | not | `a` is false (it flips the value) |

`&&` stops at the first `false`, and `||` stops at the first `true`: the rest isn't even evaluated. This is called **short-circuit**, and it protects tests that would fail, such as dividing by zero.

````java
int age = 20;
boolean student = true;
IO.println(age >= 18 && student);   // both are true
IO.println(age < 12 || student);    // one is enough
IO.println(!student);               // flipped

int count = 0;
int total = 50;
if (count != 0 && total / count > 5) {
    IO.println("average above 5");
} else {
    IO.println("no data, or a small average");
}
````

## 6. Comparing text: equals, not ==

For `String`s, `==` asks whether both variables point to **the same object** in memory, not whether they have the same letters. Two texts with equal letters can be different objects, as happens with text typed by the user. To compare the letters, use `equals`, or `equalsIgnoreCase` to ignore upper and lower case.

````java
String answer = "yes";
String typed = new String("yes");   // a separate object, like text read from the keyboard
IO.println(typed == answer);        // same object?
IO.println(typed.equals(answer));   // same letters?
IO.println("YES".equalsIgnoreCase(answer));

if (typed.equals("yes")) {
    IO.println("Confirmed");
}
````

## 7. switch: choosing by value

When you compare **one value** against a list of fixed options, `switch` is clearer than a long `else if` chain. Each `case` lists one or more values, `->` gives the result, and `default` covers everything else:

```java
String kind = switch (day) {
    case 6, 7 -> "weekend";
    default -> "weekday";
};
```

A `switch` like this one is an **expression**: it produces a value, so it must cover every possible input (that's what `default` is for) and ends with `;`.

````java
int day = 6;   // 1 = Monday ... 7 = Sunday
String kind = switch (day) {
    case 1, 2, 3, 4, 5 -> "weekday";
    case 6, 7 -> "weekend";
    default -> "not a day";
};
IO.println(day + " is a " + kind);

String command = "stop";
switch (command) {   // a switch can also just run code, and works with Strings
    case "go" -> IO.println("Moving");
    case "stop" -> IO.println("Stopped");
    default -> IO.println("Unknown command");
}
````

## 8. The ternary operator: a short if / else

`condition ? valueIfTrue : valueIfFalse` picks one of two **values**. It's a short `if` / `else` that fits inside an expression, handy for small choices such as singular or plural. For anything longer, a normal `if` reads better.

````java
int items = 1;
String word = items == 1 ? "item" : "items";
IO.println(items + " " + word);

int score = 72;
IO.println("Result: " + (score >= 60 ? "pass" : "fail"));
````

## 9. Common mistakes

- **A `;` right after `if (…)`** ends the `if` there: the block below becomes an ordinary block that always runs. Run the cell below and see.
- **`=` instead of `==`** in a condition: `if (age = 18)` doesn't even compile, because `age = 18` is an `int`, not a `boolean`.
- **`==` with Strings:** use `equals` (step 6).
- **Conditions in the wrong order:** the specific case first (step 4).

````java
int x = 5;
if (x > 10); {
    IO.println("x is big?");   // printed anyway: the ; ended the if
}
````

🧪 **Try it:** remove the `;` after `if (x > 10)` and run again.

## In short

`if` runs a block only when its condition is `true`. Conditions compare values with `==`, `!=`, `<`, `<=`, `>` and `>=`, and combine them with `&&` (and), `||` (or) and `!` (not).

```java
void main() {
    int age = Integer.parseInt(IO.readln());
    int day = Integer.parseInt(IO.readln()); // 1 = Monday … 7 = Sunday
    if (age < 12) {
        IO.println("Child ticket");
    } else if (age >= 65 || day == 3) {
        IO.println("Discount ticket");
    } else {
        IO.println("Full ticket");
    }
    String kind = switch (day) {
        case 6, 7 -> "weekend";
        default -> "weekday";
    };
    IO.println(age >= 18 ? "Adult on a " + kind : "Minor on a " + kind);
}
```

Java tests `if`, then each `else if`, from top to bottom, and runs only the **first** block whose condition is true.

**Watch out:** the order matters. With `if (age >= 18)` before `if (age >= 65)`, the second test is never reached.
