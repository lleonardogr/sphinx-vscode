## 1. Por que métodos?

Um **método** é um trecho de código com nome. Você o escreve uma vez e o **chama** pelo nome quantas vezes quiser. Métodos evitam repetir código e dividem um programa grande em partes pequenas, fáceis de ler e testar. Você já usa um: o `main` é o método onde o programa começa.

Nas células daqui em diante, os métodos ficam escritos acima de `void main()`, e o `main` os chama.

````java
void printLine() {
    IO.println("--------------------");
}

void main() {
    printLine();
    IO.println("  Relatório semanal");
    printLine();
    IO.println("  3 tarefas feitas");
    printLine();
}
````

🧪 **Experimente:** troque os traços por `=` em **um** lugar só e rode: todas as linhas mudam. Depois chame `printLine()` mais uma vez no fim.

## 2. Parâmetros: o que o método precisa

Entre os parênteses, um método lista os **parâmetros**: variáveis, cada uma com um tipo, que recebem valores quando ele é chamado. Os valores na chamada são os **argumentos**, e vão na mesma ordem: `printLine(10, '*')` põe 10 em `width` e `'*'` em `symbol`.

````java
void printLine(int width, char symbol) {
    IO.println(String.valueOf(symbol).repeat(width));
}

void greet(String name, int times) {
    for (int i = 0; i < times; i++) {
        IO.println("Oi, " + name + "!");
    }
}

void main() {
    printLine(10, '*');
    greet("Ada", 2);
    printLine(20, '-');
}
````

🧪 **Experimente:** chame `greet(2, "Ada")` com os argumentos trocados e leia o erro. Depois chame `printLine` com um argumento só.

## 3. return: devolvendo uma resposta

Um método pode calcular uma resposta e **devolvê-la** a quem o chamou. O tipo dele, escrito antes do nome, diz que tipo de valor volta: `double discounted(...)` devolve um `double`. `return valor;` encerra o método e entrega o valor, e a chamada é substituída por ele: `discounted(100, 10)` **vira** `90.0`.

Os slides mostram a pilha de chamadas: uma chamada empilha um quadro com cópias dos argumentos, e o `return` o desempilha.

#### 🎞️ Uma chamada de método na pilha de chamadas

![Chamar discounted(100, 10) empilha um quadro com as próprias cópias dos argumentos; o return o desempilha e devolve 90 para o main](call-stack-steps.pt-br.svg)

````java
double discounted(double price, int percent) {
    return price - price * percent / 100;
}

void main() {
    double price = 100;
    double sale = discounted(price, 10);       // guarda a resposta
    IO.println("10% de desconto: " + sale);
    IO.println("25% de desconto: " + discounted(price, 25));   // ou usa na hora
    discounted(price, 50);                     // calculado e jogado fora!
}
````

🧪 **Experimente:** imprima também o desconto de 50%, usando o resultado da última chamada. Depois aplique dois descontos seguidos: `discounted(discounted(price, 10), 10)`. Dá o mesmo que 20%?

## 4. void, e saindo antes

Um método que só **faz** algo (imprime, muda coisas) e não devolve nada tem o tipo `void`. Um método de qualquer outro tipo **precisa** devolver um valor em todo caminho, senão o código não compila. O `return` também pode sair do método antes, assim que a resposta é conhecida.

````java
boolean isAdult(int age) {
    return age >= 18;          // uma resposta boolean
}

String category(int age) {
    if (age < 0) {
        return "inválida";     // sai aqui mesmo
    }
    if (age < 18) {
        return "menor";
    }
    return "adulto";
}

void main() {
    IO.println(isAdult(20));
    IO.println(category(-3) + ", " + category(10) + ", " + category(40));
}
````

🧪 **Experimente:** apague o último `return "adulto";` e leia o erro ("missing return statement"): um caminho fica sem resposta. Volte com ele.

## 5. Cada método tem as próprias variáveis

Variáveis criadas dentro de um método só existem ali, enquanto ele roda. Um parâmetro é uma **cópia** do valor do argumento: mudá-lo dentro do método não muda a variável de quem chamou. Para tirar um valor de lá, devolva-o com `return`.

````java
void tryToDouble(int n) {
    n = n * 2;
    IO.println("dentro: " + n);
}

int doubled(int n) {
    return n * 2;
}

void main() {
    int n = 5;
    tryToDouble(n);
    IO.println("depois de tryToDouble: " + n);
    n = doubled(n);
    IO.println("depois de n = doubled(n): " + n);
}
````

🧪 **Experimente:** preveja as três linhas antes. Depois tente imprimir, no `main`, uma variável criada dentro de `tryToDouble`: o compilador não a conhece.

## 6. Métodos que chamam métodos

Um método pode chamar outros métodos. Faça métodos pequenos que fazem uma coisa cada, depois combine-os: o método maior fica quase como uma frase.

````java
double circleArea(double radius) {
    return Math.PI * radius * radius;
}

double ringArea(double outer, double inner) {
    return circleArea(outer) - circleArea(inner);
}

void main() {
    IO.println(circleArea(1));
    IO.println(ringArea(3, 2));
}
````

🧪 **Experimente:** escreva `double squareArea(double side)` e use-o num método para a área de uma moldura quadrada (lado de fora menos lado de dentro).

## 7. Sobrecarga: mesmo nome, parâmetros diferentes

Dois métodos podem ter o mesmo nome se os parâmetros forem diferentes em quantidade ou em tipo. O Java escolhe o que combina com a chamada. É por isso que `IO.println` aceita um `int`, uma `String` ou um `double`: existem vários métodos `println`.

````java
String describe(int n) {
    return "o número " + n;
}

String describe(String s) {
    return "o texto \"" + s + "\"";
}

String describe(int n, String unit) {
    return n + " " + unit;
}

void main() {
    IO.println(describe(7));
    IO.println(describe("sete"));
    IO.println(describe(7, "dias"));
}
````

🧪 **Experimente:** chame `describe(7.5)`: nenhuma versão recebe `double`. Acrescente uma.

## Em resumo

Um **método** é um trecho de código com nome que você pode chamar quantas vezes quiser. Os **parâmetros** dele são os valores de que ele precisa; `return` devolve a resposta a quem o chamou. Um método `void` não devolve nada: ele só faz algo, como imprimir.

```java
double discounted(double price, int percent) {
    return price - price * percent / 100;
}

void printLine(int width) {
    IO.println("-".repeat(width));
}

void main() {
    double price = Double.parseDouble(IO.readln());
    printLine(20);
    IO.println("10% de desconto: " + discounted(price, 10));
    IO.println("25% de desconto: " + discounted(price, 25));
    printLine(20);
}
```

Cada método tem suas próprias variáveis: o `price` dentro de `discounted` é uma cópia do valor com que ele foi chamado.

**Cuidado:** `discounted(price, 10);` sozinho numa linha calcula a resposta e a joga fora. Use o resultado: imprima ou guarde, como em `double sale = discounted(price, 10);`.
