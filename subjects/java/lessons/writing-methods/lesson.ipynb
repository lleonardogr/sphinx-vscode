## 1. Why methods?

A **method** is a piece of code with a name. You write it once and **call** it by name as often as you like. Methods avoid repeating code, and they split a big program into small parts that are easy to read and test. You already use one: `main` is the method where the program starts.

In the cells from now on, the methods are written above `void main()`, and `main` calls them.

````java
void printLine() {
    IO.println("--------------------");
}

void main() {
    printLine();
    IO.println("  Weekly report");
    printLine();
    IO.println("  3 tasks done");
    printLine();
}
````

🧪 **Try it:** change the dashes to `=` in **one** place and run: every line changes. Then call `printLine()` one more time at the end.

## 2. Parameters: what the method needs

Between the parentheses, a method lists its **parameters**: variables, each with a type, that receive values when it's called. The values in the call are the **arguments**, and they go in the same order: `printLine(10, '*')` puts 10 in `width` and `'*'` in `symbol`.

````java
void printLine(int width, char symbol) {
    IO.println(String.valueOf(symbol).repeat(width));
}

void greet(String name, int times) {
    for (int i = 0; i < times; i++) {
        IO.println("Hi, " + name + "!");
    }
}

void main() {
    printLine(10, '*');
    greet("Ada", 2);
    printLine(20, '-');
}
````

🧪 **Try it:** call `greet(2, "Ada")` with the arguments swapped and read the error. Then call `printLine` with only one argument.

## 3. return: sending an answer back

A method can compute an answer and **return** it to whoever called it. Its type, written before the name, says what kind of value comes back: `double discounted(...)` returns a `double`. `return value;` ends the method and hands the value back, and the call is replaced by it: `discounted(100, 10)` **becomes** `90.0`.

The slides show the call stack: a call pushes a frame with copies of the arguments, and `return` pops it.

#### 🎞️ A method call on the call stack

![Calling discounted(100, 10) pushes a frame with its own copies of the arguments; return pops it and hands 90 back to main](call-stack-steps.svg)

````java
double discounted(double price, int percent) {
    return price - price * percent / 100;
}

void main() {
    double price = 100;
    double sale = discounted(price, 10);       // store the answer
    IO.println("10% off: " + sale);
    IO.println("25% off: " + discounted(price, 25));   // or use it right away
    discounted(price, 50);                     // computed and thrown away!
}
````

🧪 **Try it:** print the 50% discount too, by using the result of the last call. Then apply two discounts in a row: `discounted(discounted(price, 10), 10)`. Is it the same as 20%?

## 4. void, and returning early

A method that only **does** something (prints, changes things) and gives nothing back has the type `void`. A method with any other type **must** return a value on every path, or the code doesn't compile. `return` can also leave a method early, as soon as the answer is known.

````java
boolean isAdult(int age) {
    return age >= 18;          // a boolean answer
}

String category(int age) {
    if (age < 0) {
        return "invalid";      // leaves right here
    }
    if (age < 18) {
        return "minor";
    }
    return "adult";
}

void main() {
    IO.println(isAdult(20));
    IO.println(category(-3) + ", " + category(10) + ", " + category(40));
}
````

🧪 **Try it:** delete the last `return "adult";` and read the error ("missing return statement"): one path has no answer. Put it back.

## 5. Each method has its own variables

Variables created inside a method exist only there, while it runs. A parameter is a **copy** of the argument's value: changing it inside the method doesn't change the caller's variable. To get a value out, `return` it.

````java
void tryToDouble(int n) {
    n = n * 2;
    IO.println("inside: " + n);
}

int doubled(int n) {
    return n * 2;
}

void main() {
    int n = 5;
    tryToDouble(n);
    IO.println("after tryToDouble: " + n);
    n = doubled(n);
    IO.println("after n = doubled(n): " + n);
}
````

🧪 **Try it:** predict the three lines first. Then try to print, in `main`, a variable created inside `tryToDouble`: the compiler doesn't know it.

## 6. Methods that call methods

A method can call other methods. Build small ones that do one thing each, then combine them: the big method reads almost like a sentence.

````java
double circleArea(double radius) {
    return Math.PI * radius * radius;
}

double ringArea(double outer, double inner) {
    return circleArea(outer) - circleArea(inner);
}

void main() {
    IO.println(circleArea(1));
    IO.println(ringArea(3, 2));
}
````

🧪 **Try it:** write `double squareArea(double side)` and use it in a method for the area of a square frame (outer side minus inner side).

## 7. Overloading: same name, different parameters

Two methods can share a name if their parameters differ in number or types. Java picks the one that matches the call. That's why `IO.println` accepts an `int`, a `String` or a `double`: there are several `println` methods.

````java
String describe(int n) {
    return "the number " + n;
}

String describe(String s) {
    return "the text \"" + s + "\"";
}

String describe(int n, String unit) {
    return n + " " + unit;
}

void main() {
    IO.println(describe(7));
    IO.println(describe("seven"));
    IO.println(describe(7, "days"));
}
````

String repeatWord(String word, int times, String sep) {
    String result = "";

## In short

A **method** is a named piece of code you can call as often as you like. Its **parameters** are the values it needs; `return` sends the answer back to whoever called it. A `void` method returns nothing: it just does something, such as printing.

```java
double discounted(double price, int percent) {
    return price - price * percent / 100;
}

void printLine(int width) {
    IO.println("-".repeat(width));
}

void main() {
    double price = Double.parseDouble(IO.readln());
    printLine(20);
    IO.println("10% off: " + discounted(price, 10));
    IO.println("25% off: " + discounted(price, 25));
    printLine(20);
}
```

Each method has its own variables: `price` inside `discounted` is a copy of the value it was called with.

**Watch out:** `discounted(price, 10);` alone on a line computes the answer and throws it away. Use the result: print it or store it, as in `double sale = discounted(price, 10);`.
