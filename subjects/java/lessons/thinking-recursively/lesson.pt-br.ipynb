## 1. Um método que chama a si mesmo

Um método **recursivo** chama a si mesmo, cada vez com uma versão menor do problema. Para dizer "hip" 3 vezes, diga uma vez e depois diga mais 2 vezes; para dizer 0 vezes, não faça nada. Essa última parte, pequena o bastante para responder sem outra chamada, é o **caso base**, e é ela que faz as chamadas pararem.

````java
void cheer(int times) {
    if (times == 0) {              // caso base: não sobrou nada
        IO.println("urra!");
        return;
    }
    IO.println("hip " + times);
    cheer(times - 1);              // o mesmo problema, um a menos
}

void main() {
    cheer(3);
}
````

## 2. A receita: caso base + um problema menor

Todo método recursivo segue a mesma receita:

1. **Caso base:** a menor entrada, respondida direto (um texto vazio, 0, um item só).
2. **Passo recursivo:** suponha que o método já funciona para uma entrada menor, chame-o, e use a resposta dele para montar a sua.

Para a soma 1 + 2 + … + n: a soma até 0 é 0, e a soma até n é **n mais a soma até n − 1**.

````java
int sumTo(int n) {
    if (n == 0) {
        return 0;                    // caso base
    }
    return n + sumTo(n - 1);         // confie na chamada menor
}

void main() {
    IO.println(sumTo(4));
    IO.println(sumTo(100));
}
````

## 3. Acompanhando as chamadas na pilha

Cada chamada ganha o próprio quadro na pilha de chamadas, com o próprio `n`. As chamadas se empilham até o caso base; depois voltam uma a uma, cada uma terminando o seu trabalho com a resposta da chamada acima dela. Os slides acompanham `binary(13)`, que monta 13 em binário como `"1101"`.

#### 🎞️ Chamadas recursivas na pilha de chamadas

![binary(13) chama binary(6), binary(3) e binary(1); o caso base devolve "1" e cada chamada acrescenta um dígito na volta: "11", "110", "1101"](recursion-steps.pt-br.svg)

````java
String binary(int n, int depth) {
    String indent = "  ".repeat(depth);
    IO.println(indent + "binary(" + n + ")");
    if (n < 2) {
        IO.println(indent + "→ \"" + n + "\" (caso base)");
        return String.valueOf(n);
    }
    String result = binary(n / 2, depth + 1) + n % 2;
    IO.println(indent + "→ \"" + result + "\"");
    return result;
}

void main() {
    IO.println(binary(13, 0));
}
````

## 4. Recursão em textos e arrays

Numa String, o problema menor costuma ser **o texto sem o primeiro caractere**: `s.substring(1)`. O caso base é o texto vazio. Num array, não copie: passe um **índice** dizendo onde o resto começa, e o caso base é quando o índice chega ao fim.

````java
int vowels(String s) {
    if (s.isEmpty()) {
        return 0;
    }
    int first = "aeiou".indexOf(s.charAt(0)) >= 0 ? 1 : 0;
    return first + vowels(s.substring(1));
}

int sum(int[] a, int from) {
    if (from == a.length) {
        return 0;
    }
    return a[from] + sum(a, from + 1);
}

void main() {
    IO.println(vowels("recursao"));
    IO.println(sum(new int[] {4, 8, 15, 16}, 0));
}
````

## 5. Quando as mesmas chamadas se repetem

Alguns métodos recursivos fazem **duas** chamadas cada. Aí as chamadas se ramificam como uma árvore, e os mesmos problemas menores são resolvidos de novo e de novo: o tempo pode explodir. O conserto é a **memoização**: guarde cada resposta na primeira vez (num `Map`), e consulte antes de calcular de novo.

Abaixo, `paths(r, c)` conta os caminhos numa grade, do canto de cima à esquerda até a linha r, coluna c, andando só para a direita ou para baixo.

````java
long calls = 0;

long paths(int r, int c) {
    calls++;
    if (r == 0 || c == 0) {
        return 1;                         // uma linha reta só
    }
    return paths(r - 1, c) + paths(r, c - 1);
}

Map<String, Long> memo = new HashMap<>();

long fastPaths(int r, int c) {
    calls++;
    if (r == 0 || c == 0) {
        return 1;
    }
    String key = r + "," + c;
    if (memo.containsKey(key)) {
        return memo.get(key);             // já calculado
    }
    long result = fastPaths(r - 1, c) + fastPaths(r, c - 1);
    memo.put(key, result);
    return result;
}

void main() {
    IO.println(paths(12, 12) + " caminhos, " + calls + " chamadas");
    calls = 0;
    IO.println(fastPaths(12, 12) + " caminhos, " + calls + " chamadas");
}
````

## 6. Recursão ou laço?

Tudo o que um laço faz, a recursão também faz, e vice-versa. A recursão brilha quando o problema é naturalmente **feito de cópias menores de si mesmo**: árvores, pastas dentro de pastas, dividir uma lista ao meio. Para uma contagem ou soma simples, um laço é mais claro, e não enche a pilha de chamadas: milhares de chamadas aninhadas acabam num `StackOverflowError`.

````java
int sumLoop(int n) {
    int total = 0;
    for (int i = 1; i <= n; i++) {
        total += i;
    }
    return total;
}

int sumRecursive(int n) {
    return n == 0 ? 0 : n + sumRecursive(n - 1);
}

void main() {
    IO.println(sumLoop(1000) + " " + sumRecursive(1000));
}
````

🧪 **Experimente:** tente `100000` nos dois: o laço vai bem, enquanto a recursão estoura a pilha.

## Em resumo

Um método **recursivo** chama a si mesmo numa versão menor do problema. Ele precisa de um **caso base**, pequeno o bastante para responder direto, e toda chamada tem que chegar mais perto dele.

```java
String binary(int n) {
    if (n < 2) {
        return String.valueOf(n);  // caso base: 0 ou 1
    }
    return binary(n / 2) + n % 2;  // um problema menor, depois mais um dígito
}

void main() {
    int n = Integer.parseInt(IO.readln());
    IO.println(n + " em binário é " + binary(n));
}
```

`binary(13)` chama `binary(6)`, que chama `binary(3)`, depois `binary(1)`: o caso base devolve `"1"`, e cada chamada acrescenta seu dígito na volta: `1101`.

**Cuidado:** sem caso base, ou com chamadas que nunca chegam a ele, o método chama a si mesmo até o programa parar com um `StackOverflowError`.
