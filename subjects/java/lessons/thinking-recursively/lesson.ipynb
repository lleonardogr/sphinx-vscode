## 1. A method that calls itself

A **recursive** method calls itself, each time on a smaller version of the problem. To say "hip" 3 times, say it once and then say it 2 more times; to say it 0 times, do nothing. That last part, small enough to answer without another call, is the **base case**, and it's what makes the calls stop.

````java
void cheer(int times) {
    if (times == 0) {              // base case: nothing left
        IO.println("hooray!");
        return;
    }
    IO.println("hip " + times);
    cheer(times - 1);              // the same problem, one smaller
}

void main() {
    cheer(3);
}
````

## 2. The recipe: base case + a smaller problem

Every recursive method follows the same recipe:

1. **Base case:** the smallest input, answered directly (an empty text, 0, a single item).
2. **Recursive step:** assume the method already works for a smaller input, call it, and use its answer to build yours.

For the sum 1 + 2 + … + n: the sum up to 0 is 0, and the sum up to n is **n plus the sum up to n − 1**.

````java
int sumTo(int n) {
    if (n == 0) {
        return 0;                    // base case
    }
    return n + sumTo(n - 1);         // trust the smaller call
}

void main() {
    IO.println(sumTo(4));
    IO.println(sumTo(100));
}
````

## 3. Following the calls on the call stack

Each call gets its own frame on the call stack, with its own `n`. The calls pile up until the base case; then they return one by one, each finishing its own work with the answer from the call above it. The slides follow `binary(13)`, which builds 13 in binary as `"1101"`.

#### 🎞️ Recursive calls on the call stack

![binary(13) calls binary(6), binary(3) and binary(1); the base case returns "1" and each call adds one digit on the way back: "11", "110", "1101"](recursion-steps.svg)

````java
String binary(int n, int depth) {
    String indent = "  ".repeat(depth);
    IO.println(indent + "binary(" + n + ")");
    if (n < 2) {
        IO.println(indent + "→ \"" + n + "\" (base case)");
        return String.valueOf(n);
    }
    String result = binary(n / 2, depth + 1) + n % 2;
    IO.println(indent + "→ \"" + result + "\"");
    return result;
}

void main() {
    IO.println(binary(13, 0));
}
````

🧪 **Try it:** trace `binary(6)` on paper before running it. How many calls does `binary(1000)` make? (Each one halves n.)

## 4. Recursion on text and arrays

For a String, the smaller problem is usually **the text without its first character**: `s.substring(1)`. The base case is the empty text. For an array, don't copy it: pass an **index** saying where the rest starts, and the base case is when the index reaches the end.

````java
int vowels(String s) {
    if (s.isEmpty()) {
        return 0;
    }
    int first = "aeiou".indexOf(s.charAt(0)) >= 0 ? 1 : 0;
    return first + vowels(s.substring(1));
}

int sum(int[] a, int from) {
    if (from == a.length) {
        return 0;
    }
    return a[from] + sum(a, from + 1);
}

void main() {
    IO.println(vowels("recursion"));
    IO.println(sum(new int[] {4, 8, 15, 16}, 0));
}
````

🧪 **Try it:** change `sum` to return the **largest** value instead: the base case and the combining step both change.

## 5. When the same calls repeat

Some recursive methods make **two** calls each. Then the calls branch like a tree, and the same smaller problems are solved again and again: the time can explode. The fix is **memoization**: store each answer the first time (in a `Map`), and look it up before computing again.

Below, `paths(r, c)` counts the ways to walk a grid from the top-left corner, moving only right or down, to row r, column c.

````java
long calls = 0;

long paths(int r, int c) {
    calls++;
    if (r == 0 || c == 0) {
        return 1;                         // a single straight line
    }
    return paths(r - 1, c) + paths(r, c - 1);
}

Map<String, Long> memo = new HashMap<>();

long fastPaths(int r, int c) {
    calls++;
    if (r == 0 || c == 0) {
        return 1;
    }
    String key = r + "," + c;
    if (memo.containsKey(key)) {
        return memo.get(key);             // already computed
    }
    long result = fastPaths(r - 1, c) + fastPaths(r, c - 1);
    memo.put(key, result);
    return result;
}

void main() {
    IO.println(paths(12, 12) + " paths, " + calls + " calls");
    calls = 0;
    IO.println(fastPaths(12, 12) + " paths, " + calls + " calls");
}
````

🧪 **Try it:** try `14, 14` and compare the number of calls. Don't go much higher with the slow version: each step up multiplies its calls.

## 6. Recursion or a loop?

Anything a loop does, recursion can do, and the other way round. Recursion shines when the problem is naturally **made of smaller copies of itself**: trees, folders inside folders, splitting a list in halves. For a simple count or a sum, a loop is clearer, and it doesn't fill the call stack: thousands of nested calls end in a `StackOverflowError`.

````java
long sumLoop(int n) {
    long total = 0;          // long: the sum up to 100,000 doesn't fit in an int
    for (int i = 1; i <= n; i++) {
        total += i;
    }
    return total;
}

long sumRecursive(int n) {
    return n == 0 ? 0 : n + sumRecursive(n - 1);
}

void main() {
    IO.println(sumLoop(1000) + " " + sumRecursive(1000));
}
````

🧪 **Try it:** try `100000` in both: the loop is fine, while the recursion overflows the stack.

## In short

A **recursive** method calls itself on a smaller version of the problem. It needs a **base case**, small enough to answer directly, and every call must get closer to it.

```java
String binary(int n) {
    if (n < 2) {
        return String.valueOf(n);  // base case: 0 or 1
    }
    return binary(n / 2) + n % 2;  // a smaller problem, then one more digit
}

void main() {
    int n = Integer.parseInt(IO.readln());
    IO.println(n + " in binary is " + binary(n));
}
```

`binary(13)` calls `binary(6)`, which calls `binary(3)`, then `binary(1)`: the base case returns `"1"`, and each call adds its digit on the way back: `1101`.

**Watch out:** without a base case, or with calls that never reach it, the method calls itself until the program stops with a `StackOverflowError`.
