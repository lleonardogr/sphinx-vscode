## 1. An array: many values under one name

To keep the steps of 7 days you could create 7 variables, but not 365. An **array** holds many values of **one type** under one name. Its size is fixed when it's created:

```java
int[] steps = new int[7];          // 7 ints, all starting at 0
int[] ages = {15, 17, 16};         // created with these values
```

`steps.length` is the size (no parentheses, unlike a String's `length()`). `Arrays.toString(steps)` turns the array into readable text.

````java
int[] steps = new int[7];
int[] ages = {15, 17, 16};
String[] names = {"Ana", "Bia", "Caio"};
IO.println(Arrays.toString(steps) + " has " + steps.length + " values");
IO.println(Arrays.toString(ages));
IO.println(Arrays.toString(names));
IO.println(ages);   // printing the array itself: just a code, not the values
````

## 2. Indexes: reading and changing one value

Each value has an **index**, from **0** to `length - 1`. `ages[0]` is the first value, and `ages[ages.length - 1]` the last. Use it like a variable: read it, or put a new value in it with `=`. An index outside that range stops the program with an `ArrayIndexOutOfBoundsException`.

````java
int[] ages = {15, 17, 16};
IO.println("First: " + ages[0]);
IO.println("Last: " + ages[ages.length - 1]);
ages[1] = 18;          // change the second value
ages[2]++;             // add 1 to the third
IO.println(Arrays.toString(ages));
````

🧪 **Try it:** print `ages[3]` and read the exception: which index did it try, and what is the length? Then try `ages[-1]`.

## 3. Going through an array

A `for` loop with `i` from 0 to `length - 1` visits every index, so you can read or change each value. When you only need to **read** the values, the **for-each** loop is shorter: `for (int a : ages)` gives each value in turn, with no index.

````java
int[] ages = {15, 17, 16, 14};
for (int i = 0; i < ages.length; i++) {
    IO.println("index " + i + ": " + ages[i]);
}
for (int i = 0; i < ages.length; i++) {
    ages[i] = ages[i] + 1;     // everyone has a birthday
}
for (int a : ages) {
    IO.print(a + " ");
}
IO.println();
````

🧪 **Try it:** change `i < ages.length` to `i <= ages.length` and run: the last round goes one past the end. Then try to add 1 inside the for-each (`a++`): does the array change?

## 4. Neighbours and flags

Two patterns come up all the time. **Neighbours:** compare `a[i]` with `a[i - 1]`, starting at `i = 1` so `i - 1` exists. **A flag:** a `boolean` that starts `false` and turns `true` when you find what you're looking for.

````java
int[] steps = {4000, 6500, 6200, 9000, 12000};
for (int i = 1; i < steps.length; i++) {
    int change = steps[i] - steps[i - 1];
    IO.println("Day " + (i + 1) + ": " + (change >= 0 ? "+" : "") + change);
}

boolean over10k = false;
for (int s : steps) {
    if (s > 10000) {
        over10k = true;
    }
}
IO.println("Any day over 10,000? " + over10k);
````

## 5. Arrays are references

An array lives in the heap, and the variable holds an arrow to it. So `int[] b = a;` doesn't copy the values: `a` and `b` are **the same array**, and a change through one shows through the other. `a.clone()` makes a real copy. For the same reason, `==` asks whether two arrays are the same object; `Arrays.equals(a, b)` compares the values.

#### 🎞️ Arrays and references

![An array lives in the heap; assigning it to another variable shares it, while clone makes a separate copy](array-memory-steps.svg)

````java
int[] a = new int[3];
a[1] = 7;
int[] b = a;           // the same array
b[0] = 9;
int[] c = a.clone();   // a separate copy
c[2] = 5;
IO.println("a = " + Arrays.toString(a));
IO.println("b = " + Arrays.toString(b));
IO.println("c = " + Arrays.toString(c));
IO.println(a == b);
IO.println(a == a.clone());
IO.println(Arrays.equals(a, a.clone()));
````

## 6. Ready-made helpers: Arrays

The `Arrays` class has methods for common jobs: `Arrays.sort(a)` puts the values in order (it changes the array), `Arrays.fill(a, x)` fills every box with `x`, and `Arrays.copyOf(a, n)` makes a copy with a new length. Some challenges ask you to write these yourself; you can still use the library versions to check your answers.

````java
int[] scores = {72, 95, 60, 88};
Arrays.sort(scores);
IO.println(Arrays.toString(scores));
int[] bigger = Arrays.copyOf(scores, 6);   // the new boxes start at 0
IO.println(Arrays.toString(bigger));
char[] row = new char[5];
Arrays.fill(row, '#');
IO.println(new String(row));
````

🧪 **Try it:** sort a `String[]` of names: what order does it use? Then use `Arrays.copyOf` with a smaller length.

## 7. 2D arrays: rows and columns

A 2D array is an **array of rows**: `int[][] seats = new int[3][4];` has 3 rows of 4 columns. `seats[row][col]` reaches one box, `seats.length` is the number of rows, and `seats[r].length` the number of columns in row `r`. Two nested loops go through all of it.

````java
int[][] seats = new int[3][4];   // 0 = free, 1 = taken
seats[0][1] = 1;
seats[2][3] = 1;
for (int r = 0; r < seats.length; r++) {
    for (int c = 0; c < seats[r].length; c++) {
        IO.print(seats[r][c] == 1 ? "X " : ". ");
    }
    IO.println();
}
````

🧪 **Try it:** take the whole second row (`seats[1][c]` for every c). Then print `Arrays.toString(seats[0])`: one row is an ordinary array.

## In short

An **array** holds a fixed number of values of one type, numbered from 0 to `length - 1`. You read or change one with its **index**: `steps[0]` is the first.

```java
void main() {
    int n = Integer.parseInt(IO.readln());
    String[] parts = IO.readln().trim().split(" ");
    int[] steps = new int[n];
    for (int i = 0; i < n; i++) {
        steps[i] = Integer.parseInt(parts[i]);
    }
    for (int i = 1; i < steps.length; i++) {
        IO.println("Day " + (i + 1) + ": " + (steps[i] - steps[i - 1]) + " vs the day before");
    }
    for (int s : steps) {
        IO.println("*".repeat(s / 1000));
    }
    int[][] seats = new int[3][4]; // 3 rows of 4, all 0
    seats[1][2] = 1;
}
```

`for (int s : steps)` visits each value when you don't need the index. A 2D array is an array of rows: `seats[row][column]`.

**Watch out:** the last index is `length - 1`. `steps[steps.length]` stops the program with an `ArrayIndexOutOfBoundsException`.
