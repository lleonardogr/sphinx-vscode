## In short

An **array** holds a fixed number of values of one type, numbered from 0 to `length - 1`. You read or change one with its **index**: `steps[0]` is the first.

#### 🎞️ Arrays and references

![An array lives in the heap; assigning it to another variable shares it, while clone makes a separate copy](array-memory-steps.svg)

```java
void main() {
    int n = Integer.parseInt(IO.readln());
    String[] parts = IO.readln().trim().split(" ");
    int[] steps = new int[n];
    for (int i = 0; i < n; i++) {
        steps[i] = Integer.parseInt(parts[i]);
    }
    for (int i = 1; i < steps.length; i++) {
        IO.println("Day " + (i + 1) + ": " + (steps[i] - steps[i - 1]) + " vs the day before");
    }
    for (int s : steps) {
        IO.println("*".repeat(s / 1000));
    }
    int[][] seats = new int[3][4]; // 3 rows of 4, all 0
    seats[1][2] = 1;
}
```

`for (int s : steps)` visits each value when you don't need the index. A 2D array is an array of rows: `seats[row][column]`.

**Predict, then run:** what does each line print? Two variables can point to the same array.

````java
int[] a = new int[4];
IO.println(Arrays.toString(a));        // new arrays start at 0
a[1] = 7;
int[] b = a;                           // b is the same array, not a copy
b[0] = 99;
IO.println(Arrays.toString(a));
int[] c = a.clone();                   // a real copy
c[3] = -1;
IO.println(Arrays.toString(a) + " " + Arrays.toString(c));
int[][] grid = {{1, 2, 3}, {4, 5, 6}};
IO.println(grid.length + " rows, " + grid[0].length + " columns: " + Arrays.deepToString(grid));
````

/** Doubles an array, then shows the original: it must not have changed. */
String originalAfterDoubling() {
    int[] original = {1, 2, 3};
    doubled(original);
    return Arrays.toString(original);
}

int[] doubled(int[] values) {
    int[] result = new int[values.length];
    for (int i = 0; i < values.length; i++) {
        result[i] = values[i] * 2;
    }
    return result;
}

void main() {
    check("doubled", Arrays.toString(doubled(new int[] {1, 2, 3})), "[2, 4, 6]");
