## 1. Um array: muitos valores com um nome só

Para guardar os passos de 7 dias você poderia criar 7 variáveis, mas não 365. Um **array** guarda muitos valores de **um tipo** com um nome só. O tamanho dele é fixo, definido na criação:

```java
int[] steps = new int[7];          // 7 ints, todos começando em 0
int[] ages = {15, 17, 16};         // criado com estes valores
```

`steps.length` é o tamanho (sem parênteses, ao contrário do `length()` de uma String). `Arrays.toString(steps)` transforma o array num texto legível.

````java
int[] steps = new int[7];
int[] ages = {15, 17, 16};
String[] names = {"Ana", "Bia", "Caio"};
IO.println(Arrays.toString(steps) + " tem " + steps.length + " valores");
IO.println(Arrays.toString(ages));
IO.println(Arrays.toString(names));
IO.println(ages);   // imprimir o array em si: só um código, não os valores
````

## 2. Índices: lendo e mudando um valor

Cada valor tem um **índice**, de **0** até `length - 1`. `ages[0]` é o primeiro valor, e `ages[ages.length - 1]` o último. Use-o como uma variável: leia, ou ponha um valor novo com `=`. Um índice fora dessa faixa para o programa com uma `ArrayIndexOutOfBoundsException`.

````java
int[] ages = {15, 17, 16};
IO.println("Primeiro: " + ages[0]);
IO.println("Último: " + ages[ages.length - 1]);
ages[1] = 18;          // muda o segundo valor
ages[2]++;             // soma 1 ao terceiro
IO.println(Arrays.toString(ages));
````

🧪 **Experimente:** imprima `ages[3]` e leia a exceção: qual índice ele tentou, e qual é o tamanho? Depois tente `ages[-1]`.

## 3. Percorrendo um array

Um laço `for` com `i` de 0 até `length - 1` visita cada índice, então você pode ler ou mudar cada valor. Quando só precisa **ler** os valores, o **for-each** é mais curto: `for (int a : ages)` entrega cada valor em sequência, sem índice.

````java
int[] ages = {15, 17, 16, 14};
for (int i = 0; i < ages.length; i++) {
    IO.println("índice " + i + ": " + ages[i]);
}
for (int i = 0; i < ages.length; i++) {
    ages[i] = ages[i] + 1;     // todo mundo faz aniversário
}
for (int a : ages) {
    IO.print(a + " ");
}
IO.println();
````

## 4. Vizinhos e sinalizadores

Dois padrões aparecem o tempo todo. **Vizinhos:** compare `a[i]` com `a[i - 1]`, começando em `i = 1` para `i - 1` existir. **Um sinalizador:** um `boolean` que começa `false` e vira `true` quando você acha o que procura.

````java
int[] steps = {4000, 6500, 6200, 9000, 12000};
for (int i = 1; i < steps.length; i++) {
    int change = steps[i] - steps[i - 1];
    IO.println("Dia " + (i + 1) + ": " + (change >= 0 ? "+" : "") + change);
}

boolean over10k = false;
for (int s : steps) {
    if (s > 10000) {
        over10k = true;
    }
}
IO.println("Algum dia acima de 10.000? " + over10k);
````

## 5. Arrays são referências

Um array mora no heap, e a variável guarda uma seta para ele. Então `int[] b = a;` não copia os valores: `a` e `b` são **o mesmo array**, e uma mudança por um aparece pelo outro. `a.clone()` faz uma cópia de verdade. Pelo mesmo motivo, `==` pergunta se dois arrays são o mesmo objeto; `Arrays.equals(a, b)` compara os valores.

#### 🎞️ Arrays e referências

![Um array mora no heap; atribuí-lo a outra variável o compartilha, enquanto clone faz uma cópia separada](array-memory-steps.pt-br.svg)

````java
int[] a = new int[3];
a[1] = 7;
int[] b = a;           // o mesmo array
b[0] = 9;
int[] c = a.clone();   // uma cópia separada
c[2] = 5;
IO.println("a = " + Arrays.toString(a));
IO.println("b = " + Arrays.toString(b));
IO.println("c = " + Arrays.toString(c));
IO.println(a == b);
IO.println(a == a.clone());
IO.println(Arrays.equals(a, a.clone()));
````

## 6. Ajudantes prontos: Arrays

A classe `Arrays` tem métodos para tarefas comuns: `Arrays.sort(a)` põe os valores em ordem (ele muda o array), `Arrays.fill(a, x)` preenche cada caixa com `x`, e `Arrays.copyOf(a, n)` faz uma cópia com outro tamanho. Alguns desafios pedem que você mesmo escreva isso; ainda assim dá para usar as versões da biblioteca para conferir as suas respostas.

## 7. Arrays 2D: linhas e colunas

Um array 2D é um **array de linhas**: `int[][] seats = new int[3][4];` tem 3 linhas de 4 colunas. `seats[linha][coluna]` alcança uma caixa, `seats.length` é o número de linhas, e `seats[r].length` o número de colunas da linha `r`. Dois laços aninhados percorrem tudo.

````java
int[][] seats = new int[3][4];   // 0 = livre, 1 = ocupado
seats[0][1] = 1;
seats[2][3] = 1;
for (int r = 0; r < seats.length; r++) {
    for (int c = 0; c < seats[r].length; c++) {
        IO.print(seats[r][c] == 1 ? "X " : ". ");
    }
    IO.println();
}
````

## Em resumo

Um **array** guarda um número fixo de valores de um tipo, numerados de 0 a `length - 1`. Você lê ou muda um deles pelo **índice**: `steps[0]` é o primeiro.

```java
void main() {
    int n = Integer.parseInt(IO.readln());
    String[] parts = IO.readln().trim().split(" ");
    int[] steps = new int[n];
    for (int i = 0; i < n; i++) {
        steps[i] = Integer.parseInt(parts[i]);
    }
    for (int i = 1; i < steps.length; i++) {
        IO.println("Dia " + (i + 1) + ": " + (steps[i] - steps[i - 1]) + " em relação ao dia anterior");
    }
    for (int s : steps) {
        IO.println("*".repeat(s / 1000));
    }
    int[][] seats = new int[3][4]; // 3 fileiras de 4, todas 0
    seats[1][2] = 1;
}
```

`for (int s : steps)` visita cada valor quando você não precisa do índice. Um array 2D é um array de linhas: `seats[linha][coluna]`.

**Cuidado:** o último índice é `length - 1`. `steps[steps.length]` para o programa com uma `ArrayIndexOutOfBoundsException`.
