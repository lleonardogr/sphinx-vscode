## Em resumo

Um **array** guarda um número fixo de valores de um tipo, numerados de 0 a `length - 1`. Você lê ou muda um deles pelo **índice**: `steps[0]` é o primeiro.

#### 🎞️ Arrays e referências

![Um array mora no heap; atribuí-lo a outra variável o compartilha, enquanto clone faz uma cópia separada](array-memory-steps.pt-br.svg)

```java
void main() {
    int n = Integer.parseInt(IO.readln());
    String[] parts = IO.readln().trim().split(" ");
    int[] steps = new int[n];
    for (int i = 0; i < n; i++) {
        steps[i] = Integer.parseInt(parts[i]);
    }
    for (int i = 1; i < steps.length; i++) {
        IO.println("Dia " + (i + 1) + ": " + (steps[i] - steps[i - 1]) + " em relação ao dia anterior");
    }
    for (int s : steps) {
        IO.println("*".repeat(s / 1000));
    }
    int[][] seats = new int[3][4]; // 3 fileiras de 4, todas 0
    seats[1][2] = 1;
}
```

`for (int s : steps)` visita cada valor quando você não precisa do índice. Um array 2D é um array de linhas: `seats[linha][coluna]`.

**Preveja e depois rode:** o que cada linha imprime? Duas variáveis podem apontar para o mesmo array.

````java
int[] a = new int[4];
IO.println(Arrays.toString(a));        // arrays novos começam com 0
a[1] = 7;
int[] b = a;                           // b é o mesmo array, não uma cópia
b[0] = 99;
IO.println(Arrays.toString(a));
int[] c = a.clone();                   // uma cópia de verdade
c[3] = -1;
IO.println(Arrays.toString(a) + " " + Arrays.toString(c));
int[][] grid = {{1, 2, 3}, {4, 5, 6}};
IO.println(grid.length + " linhas, " + grid[0].length + " colunas: " + Arrays.deepToString(grid));
````

String originalAfterDoubling() {
    int[] original = {1, 2, 3};
    doubled(original);
    return Arrays.toString(original);

int[] doubled(int[] values) {
    int[] result = new int[values.length];
