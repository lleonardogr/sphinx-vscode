## 1. Por que laços?

Para imprimir "Olá" 5 vezes você poderia escrever 5 linhas, mas não 1.000 linhas, nem "quantas vezes o usuário pedir". Um **laço** repete um bloco de código para você. O Java tem três: `for` (um número conhecido de vezes), `while` (enquanto uma condição for verdadeira) e `do … while` (pelo menos uma vez).

````java
for (int i = 1; i <= 5; i++) {
    IO.println("Olá número " + i);
}
````

## 2. O laço for, parte por parte

Um `for` tem três partes entre os parênteses, separadas por `;`:

```java
for (int i = 1; i <= 3; i++) { … }
//   início     condição   atualização
```

1. **início** roda uma vez, antes de tudo: cria o contador `i`.
2. **condição** é testada **antes de cada volta**: se for `false`, o laço acaba.
3. **atualização** roda **depois de cada volta**: `i++` soma 1.

Os slides acompanham um laço somando os quadrados de 1 a 3, volta por volta.

#### 🎞️ Um laço for, uma volta de cada vez

![Um laço for somando os quadrados de 1 a 3: total vai a 1, 5, 14, e o laço para quando i chega a 4](loop-steps.pt-br.svg)

````java
int total = 0;
for (int i = 1; i <= 3; i++) {
    total += i * i;
    IO.println("i = " + i + ", total = " + total);
}
IO.println("Depois do laço: " + total);
````

🧪 **Experimente:** mude `i <= 3` para `i < 3` e conte as voltas. Depois comece em `i = 0`. Um laço com `i = 0; i < n` roda exatamente n vezes.

## 3. Contando para trás e de tantos em tantos

A atualização pode ser qualquer coisa: `i--` conta para trás, `i += 2` pula de dois em dois. Só confira se a condição combina com a direção: contando para trás, a condição costuma ser `i > 0` ou `i >= 1`.

````java
for (int i = 5; i > 0; i--) {
    IO.print(i + "... ");
}
IO.println("Decolar!");

for (int i = 0; i <= 10; i += 2) {
    IO.print(i + " ");
}
IO.println();
````

## 4. Guardando um total ou uma contagem

Muitos laços **acumulam**: uma variável criada **antes** do laço é atualizada a cada volta e guarda a resposta **depois** dele. Um total começa em 0 e cresce com `+=`; uma contagem começa em 0 e cresce com `++` só quando algo é verdadeiro.

````java
int total = 0;
int multiplesOf3 = 0;
for (int i = 1; i <= 20; i++) {
    total += i;
    if (i % 3 == 0) {
        multiplesOf3++;
    }
}
IO.println("Total: " + total);
IO.println("Múltiplos de 3: " + multiplesOf3);
````

## 5. while: repetindo até algo mudar

Use `while` quando você **não sabe quantas voltas** vai levar: ele repete enquanto a condição for `true`, testando-a antes de cada volta. Algo dentro do laço precisa mudar a condição, senão ele nunca para.

````java
double savings = 100;
int years = 0;
while (savings < 200) {
    savings = savings * 1.1;   // 10% ao ano
    years++;
}
IO.println("Dobrou depois de " + years + " anos: " + savings);
````

## 6. do … while: pelo menos uma vez

`do { … } while (condição);` testa a condição **depois** do corpo, então o corpo sempre roda pelo menos uma vez. Combina com "faça algo, depois decida se repete", como perguntar de novo depois de uma resposta inválida. Repare no `;` no fim.

````java
int n = 0;
do {
    IO.println("Roda com n = " + n);
    n++;
} while (n < 0);

int m = 0;
while (m < 0) {
    IO.println("Nunca impresso");
}
````

🧪 **Experimente:** mude as duas condições para `< 3` e compare as saídas. Cuidado: o `while` não muda `m`, então acrescente `m++` dentro dele antes, senão ele nunca termina.

## 7. break e continue

Dentro de qualquer laço, `break` **sai do laço** na hora, e `continue` **pula para a próxima volta**. O `break` é útil para parar assim que você achou algo.

````java
for (int i = 50; i <= 100; i++) {
    if (i % 7 == 0) {
        IO.println("Primeiro múltiplo de 7 a partir de 50: " + i);
        break;
    }
}

for (int i = 1; i <= 10; i++) {
    if (i % 2 == 0) {
        continue;   // pula os pares
    }
    IO.print(i + " ");
}
IO.println();
````

🧪 **Experimente:** tire o `break` e rode: quantas linhas agora? Depois faça o segundo laço pular os múltiplos de 3.

## 8. Laços dentro de laços

Um laço pode ficar dentro de outro. Para **cada** volta do laço de fora, o laço de dentro roda **todas** as voltas dele. Com 3 linhas e 4 colunas, o corpo de dentro roda 3 × 4 = 12 vezes. É assim que se percorre grades e tabelas.

````java
for (int row = 1; row <= 3; row++) {
    for (int col = 1; col <= 4; col++) {
        IO.print("(" + row + "," + col + ") ");
    }
    IO.println();   // fim da linha
}
````

## 9. Erros comuns

- **Um a mais ou a menos:** `i <= n` roda uma vez a mais que `i < n`. Conte as voltas com um n pequeno.
- **Um laço que nunca acaba:** a condição nunca fica `false`, por exemplo um `while` que não muda a sua variável. Numa célula do notebook ele para depois de 20 segundos.
- **Um `;` logo depois de `for (…)` ou `while (…)`:** o laço fica com o corpo vazio, e o bloco de baixo roda uma vez.
- **O total criado dentro do laço:** ele volta a 0 a cada volta (passo 4).

````java
for (int i = 0; i < 3; i++); {
    IO.println("Impresso uma vez só: o ; era o corpo do laço");
}
````

🧪 **Experimente:** tire o `;` e rode de novo.

## Em resumo

Um laço **for** repete um bloco um número conhecido de vezes: ele começa um contador, confere uma condição antes de cada volta e atualiza o contador depois dela. Um laço **while** repete enquanto sua condição for verdadeira, quando você não sabe de antemão quantas voltas serão.

```java
void main() {
    int start = Integer.parseInt(IO.readln());
    for (int i = start; i > 0; i--) {
        IO.println(i + "...");
    }
    IO.println("Decolar!");

double savings = 100;
    int years = 0;
    while (savings < 200) {
        savings = savings * 1.1;
        years++;
    }
    IO.println("Dobrou em " + years + " anos");
}
```

`do { … } while (condição);` confere a condição depois do corpo, então o corpo sempre roda pelo menos uma vez.

**Cuidado:** `for (int i = 0; i < 5; i++)` roda 5 vezes, com `i` de 0 a 4. Com `<=` roda 6.
