## Em resumo

Um laço **for** repete um bloco um número conhecido de vezes: ele começa um contador, confere uma condição antes de cada volta e atualiza o contador depois dela. Um laço **while** repete enquanto sua condição for verdadeira, quando você não sabe de antemão quantas voltas serão.

#### 🎞️ Um laço for, uma volta de cada vez

![Um laço for somando os quadrados de 1 a 3: total vai a 1, 5, 14, e o laço para quando i chega a 4](loop-steps.pt-br.svg)

```java
void main() {
    int start = Integer.parseInt(IO.readln());
    for (int i = start; i > 0; i--) {
        IO.println(i + "...");
    }
    IO.println("Decolar!");

double savings = 100;
    int years = 0;
    while (savings < 200) {
        savings = savings * 1.1;
        years++;
    }
    IO.println("Dobrou em " + years + " anos");
}
```

`do { … } while (condição);` confere a condição depois do corpo, então o corpo sempre roda pelo menos uma vez.

**Passos e o erro de um a mais:** um laço pode contar de qualquer passo, e `<` contra `<=` muda quantas vezes ele roda.

````java
for (int i = 10; i >= 0; i -= 3) {
    IO.print(i + " ");
}
IO.println();

int withLess = 0;
for (int i = 0; i < 5; i++) {
    withLess++;
}
int withLessOrEqual = 0;
for (int i = 0; i <= 5; i++) {
    withLessOrEqual++;
}
IO.println("i < 5 roda " + withLess + " vezes, i <= 5 roda " + withLessOrEqual);

int tries = 0;
do {
    tries++;
} while (tries < 0); // falso desde o começo
IO.println("o do-while rodou " + tries + " vez");
````

String countdown(int from, int step) {
    String result = "";
    for (int i = from; i >= 1; i -= step) {
        result += (result.isEmpty() ? "" : " ") + i;
    }
    return result;
}
