## In short

A **for** loop repeats a block a known number of times: it starts a counter, checks a condition before each round, and updates the counter after it. A **while** loop repeats as long as its condition is true, when you don't know in advance how many rounds it takes.

#### 🎞️ A for loop, one turn at a time

![A for loop adding the squares of 1 to 3: total goes 1, 5, 14, and the loop stops when i reaches 4](loop-steps.svg)

```java
void main() {
    int start = Integer.parseInt(IO.readln());
    for (int i = start; i > 0; i--) {
        IO.println(i + "...");
    }
    IO.println("Liftoff!");

double savings = 100;
    int years = 0;
    while (savings < 200) {
        savings = savings * 1.1;
        years++;
    }
    IO.println("Doubled in " + years + " years");
}
```

`do { … } while (condition);` checks the condition after the body, so the body always runs at least once.

**Steps and off-by-one:** a loop can count by any step, and `<` against `<=` changes how many times it runs.

````java
for (int i = 10; i >= 0; i -= 3) {
    IO.print(i + " ");
}
IO.println();

int withLess = 0;
for (int i = 0; i < 5; i++) {
    withLess++;
}
int withLessOrEqual = 0;
for (int i = 0; i <= 5; i++) {
    withLessOrEqual++;
}
IO.println("i < 5 runs " + withLess + " times, i <= 5 runs " + withLessOrEqual);

int tries = 0;
do {
    tries++;
} while (tries < 0); // false from the start
IO.println("do-while ran " + tries + " time");
````

String countdown(int from, int step) {
    String result = "";
    for (int i = from; i >= 1; i -= step) {
        result += (result.isEmpty() ? "" : " ") + i;
    }
    return result;
}

void main() {
    check("by 3", countdown(10, 3), "10 7 4 1");
