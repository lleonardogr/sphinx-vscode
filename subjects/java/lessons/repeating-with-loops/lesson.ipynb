## 1. Why loops?

To print "Hello" 5 times you could write 5 lines, but not 1,000 lines, and not "as many times as the user asks". A **loop** repeats a block of code for you. Java has three: `for` (a known number of times), `while` (as long as a condition is true) and `do … while` (at least once).

````java
for (int i = 1; i <= 5; i++) {
    IO.println("Hello number " + i);
}
````

## 2. The for loop, piece by piece

A `for` has three parts between its parentheses, separated by `;`:

```java
for (int i = 1; i <= 3; i++) { … }
//   start      condition  update
```

1. **start** runs once, before everything: it creates the counter `i`.
2. **condition** is checked **before each round**: if it's `false`, the loop ends.
3. **update** runs **after each round**: `i++` adds 1.

The slides follow a loop adding the squares of 1 to 3, round by round.

#### 🎞️ A for loop, one turn at a time

![A for loop adding the squares of 1 to 3: total goes 1, 5, 14, and the loop stops when i reaches 4](loop-steps.svg)

````java
int total = 0;
for (int i = 1; i <= 3; i++) {
    total += i * i;
    IO.println("i = " + i + ", total = " + total);
}
IO.println("After the loop: " + total);
````

🧪 **Try it:** change `i <= 3` to `i < 3` and count the rounds. Then start at `i = 0`. A loop with `i = 0; i < n` runs exactly n times.

## 3. Counting down and in steps

The update can be anything: `i--` counts down, `i += 2` jumps by two. Just make sure the condition matches the direction: counting down, the condition is usually `i > 0` or `i >= 1`.

````java
for (int i = 5; i > 0; i--) {
    IO.print(i + "... ");
}
IO.println("Liftoff!");

for (int i = 0; i <= 10; i += 2) {
    IO.print(i + " ");
}
IO.println();
````

## 4. Keeping a total or a count

Many loops **accumulate**: a variable created **before** the loop gets updated on each round, and holds the answer **after** it. A total starts at 0 and grows with `+=`; a count starts at 0 and grows with `++` only when something is true.

````java
int total = 0;
int multiplesOf3 = 0;
for (int i = 1; i <= 20; i++) {
    total += i;
    if (i % 3 == 0) {
        multiplesOf3++;
    }
}
IO.println("Total: " + total);
IO.println("Multiples of 3: " + multiplesOf3);
````

## 5. while: repeating until something changes

Use `while` when you **don't know how many rounds** it takes: it repeats as long as its condition is `true`, checking it before each round. Something inside the loop must change the condition, or it never stops.

````java
double savings = 100;
int years = 0;
while (savings < 200) {
    savings = savings * 1.1;   // 10% a year
    years++;
}
IO.println("Doubled after " + years + " years: " + savings);
````

🧪 **Try it:** try 5% a year (`1.05`). Then start with `savings = 300`: how many rounds run now?

## 6. do … while: at least once

`do { … } while (condition);` checks the condition **after** the body, so the body always runs at least once. It fits "do something, then decide whether to repeat", such as asking again after an invalid answer. Note the `;` at the end.

````java
int n = 0;
do {
    IO.println("Runs with n = " + n);
    n++;
} while (n < 0);

int m = 0;
while (m < 0) {
    IO.println("Never printed");
}
````

🧪 **Try it:** change both conditions to `< 3` and compare the outputs. Careful: the `while` doesn't change `m`, so add `m++` inside it first, or it never ends.

## 7. break and continue

Inside any loop, `break` **leaves the loop** right away, and `continue` **skips to the next round**. `break` is handy to stop as soon as you've found something.

````java
for (int i = 50; i <= 100; i++) {
    if (i % 7 == 0) {
        IO.println("First multiple of 7 from 50: " + i);
        break;
    }
}

for (int i = 1; i <= 10; i++) {
    if (i % 2 == 0) {
        continue;   // skip the even ones
    }
    IO.print(i + " ");
}
IO.println();
````

🧪 **Try it:** remove the `break` and run: how many lines now? Then make the second loop skip the multiples of 3 instead.

## 8. Loops inside loops

A loop can sit inside another. For **each** round of the outer loop, the inner loop runs **all** its rounds. With 3 rows and 4 columns, the inner body runs 3 × 4 = 12 times. That's how you go through grids and tables.

````java
for (int row = 1; row <= 3; row++) {
    for (int col = 1; col <= 4; col++) {
        IO.print("(" + row + "," + col + ") ");
    }
    IO.println();   // end of the row
}
````

🧪 **Try it:** make a 5 × 5 square of `#`. Then make the inner loop run `col <= row` times: what shape appears?

## 9. Common mistakes

- **Off by one:** `i <= n` runs one more time than `i < n`. Count the rounds for a small n.
- **A loop that never ends:** the condition never becomes `false`, for example a `while` that doesn't change its variable. In a notebook cell it stops after 20 seconds.
- **A `;` right after `for (…)` or `while (…)`:** the loop gets an empty body, and the block below runs once.
- **The total created inside the loop:** it restarts at 0 on every round (step 4).

````java
for (int i = 0; i < 3; i++); {
    IO.println("Printed only once: the ; was the loop's body");
}
````

🧪 **Try it:** remove the `;` and run again.

## In short

A **for** loop repeats a block a known number of times: it starts a counter, checks a condition before each round, and updates the counter after it. A **while** loop repeats as long as its condition is true, when you don't know in advance how many rounds it takes.

```java
void main() {
    int start = Integer.parseInt(IO.readln());
    for (int i = start; i > 0; i--) {
        IO.println(i + "...");
    }
    IO.println("Liftoff!");

double savings = 100;
    int years = 0;
    while (savings < 200) {
        savings = savings * 1.1;
        years++;
    }
    IO.println("Doubled in " + years + " years");
}
```

`do { … } while (condition);` checks the condition after the body, so the body always runs at least once.

**Watch out:** `for (int i = 0; i < 5; i++)` runs 5 times, with `i` from 0 to 4. With `<=` it runs 6.
