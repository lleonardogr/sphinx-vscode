## In short

Collections grow and shrink as you add and remove items. A **List** keeps the order and allows repeats; a **Set** keeps one copy of each item; a **Map** links each **key** to a value.

#### 🎞️ List, Set and Map, step by step

![The same additions in a List (keeps a, b, a), a Set (keeps a, b) and a Map (a key put twice keeps the last value)](collections-steps.svg)

```java
void main() {
    Map<String, Double> prices = new HashMap<>();
    prices.put("apple", 0.5);
    prices.put("bread", 2.25);
    List<String> cart = new ArrayList<>();
    cart.add("apple");
    cart.add("bread");
    cart.add("apple");
    double total = 0;
    for (String item : cart) {
        total += prices.getOrDefault(item, 0.0);
    }
    Set<String> kinds = new HashSet<>(cart);
    IO.println(cart.size() + " items, " + kinds.size() + " different, total " + total);
}
```

`Hash…` collections are fast but unordered, `Linked…` ones keep the order items arrived in, and `Tree…` ones keep them sorted. An `ArrayDeque` works as a queue or a stack.

**The same words in four collections:** notice which keep the order, which drop repeats, and which sort.

````java
List<String> words = List.of("pear", "apple", "fig", "apple", "kiwi");
IO.println("ArrayList:     " + new ArrayList<>(words));
IO.println("HashSet:       " + new HashSet<>(words));
IO.println("LinkedHashSet: " + new LinkedHashSet<>(words));
IO.println("TreeSet:       " + new TreeSet<>(words));

List<Integer> numbers = new ArrayList<>(List.of(10, 20, 30));
numbers.remove(1);                    // removes the item at index 1
numbers.remove(Integer.valueOf(10));  // removes the value 10
IO.println(numbers);
````

List<Integer> evensOnly(List<Integer> numbers) {
    List<Integer> result = new ArrayList<>();
    for (int n : numbers) {
        if (n % 2 == 0) {
            result.add(n);
        }
    }
    return result;
}

Map<String, Integer> lengths(List<String> words) {
    Map<String, Integer> result = new HashMap<>();
    for (String w : words) {
        result.put(w, w.length());
    }
    return result;
}
