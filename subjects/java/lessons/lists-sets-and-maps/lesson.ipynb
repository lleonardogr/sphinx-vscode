## 1. A list that grows: ArrayList

An array's size is fixed. A **List** grows and shrinks as you add and remove items. You create one with the type of its items between `< >`:

```java
List<String> names = new ArrayList<>();
```

| Method | Does |
|---|---|
| `add(x)` | puts x at the end |
| `get(i)` | the item at index i |
| `set(i, x)` | replaces the item at index i |
| `remove(i)` | removes the item at index i |
| `size()` | how many items |
| `contains(x)` | whether x is in it |

Collections hold **objects**, so numbers use the class names: `List<Integer>`, `List<Double>`, never `List<int>`.

````java
List<String> names = new ArrayList<>();
names.add("Ana");
names.add("Bia");
names.add("Caio");
IO.println(names + " has " + names.size() + " names");
IO.println("Second: " + names.get(1));
names.set(0, "Alice");
names.remove(1);
IO.println(names);
IO.println(names.contains("Bia"));
````

## 2. Going through a list

A list is gone through like an array: a for-each to read every item, or a `for` with `get(i)` when you need the index. Don't remove items inside a for-each (Java stops with a `ConcurrentModificationException`); use `removeIf` instead.

````java
List<Integer> scores = new ArrayList<>(List.of(72, 95, 40, 88, 35));
int total = 0;
for (int s : scores) {
    total += s;
}
IO.println("Average: " + (double) total / scores.size());
for (int i = 0; i < scores.size(); i++) {
    IO.println((i + 1) + "º: " + scores.get(i));
}
scores.removeIf(s -> s < 50);   // removes every score below 50
IO.println(scores);
````

🧪 **Try it:** `List.of(...)` alone makes a list that can't change: try `List.of(1, 2).add(3)` and read the exception. That's why the cell wraps it in `new ArrayList<>(...)`.

## 3. Set: one copy of each item

A **Set** keeps each item at most once: adding a repeat does nothing, and `add` returns `false`. It's the tool for "the different ones". The order depends on the kind: `HashSet` has no useful order, `LinkedHashSet` keeps the order items arrived in, and `TreeSet` keeps them sorted. The slides compare a List, a Set and a Map getting the same items.

#### 🎞️ List, Set and Map, step by step

![The same additions in a List (keeps a, b, a), a Set (keeps a, b) and a Map (a key put twice keeps the last value)](collections-steps.svg)

````java
List<String> visits = List.of("home", "news", "home", "shop", "news", "home");
Set<String> pages = new LinkedHashSet<>(visits);
IO.println(visits.size() + " visits to " + pages.size() + " pages: " + pages);
IO.println(pages.add("home"));    // already there
IO.println(pages.add("help"));
IO.println(new TreeSet<>(pages)); // sorted
````

## 4. Map: looking things up by key

A **Map** links each **key** to a **value**, like a dictionary links a word to its meaning. `put(key, value)` stores (and replaces the old value if the key was there), `get(key)` reads (`null` when the key is missing), `getOrDefault(key, x)` gives `x` instead of `null`, and `containsKey` asks. To go through it, loop over `entrySet()`: each entry has `getKey()` and `getValue()`.

````java
Map<String, Double> prices = new HashMap<>();
prices.put("apple", 0.5);
prices.put("bread", 2.25);
prices.put("apple", 0.6);          // replaces 0.5
IO.println(prices.get("apple"));
IO.println(prices.get("milk"));    // not there
IO.println(prices.getOrDefault("milk", 0.0));

List<String> cart = List.of("apple", "bread", "apple", "milk");
double total = 0;
for (String item : cart) {
    total += prices.getOrDefault(item, 0.0);
}
IO.println("Total: " + total);
for (Map.Entry<String, Double> e : prices.entrySet()) {
    IO.println(e.getKey() + " costs " + e.getValue());
}
````

## 5. Queues and stacks: ArrayDeque

An `ArrayDeque` can be used from both ends. As a **queue** (first in, first out, like a line at the bank): `offer` joins at the back, `poll` takes from the front. As a **stack** (last in, first out, like a pile of plates): `push` puts on top, `pop` takes from the top. `peek` looks without removing.

````java
Deque<String> line = new ArrayDeque<>();
line.offer("Ana");
line.offer("Bia");
line.offer("Caio");
IO.println("Served: " + line.poll() + ", waiting: " + line);

Deque<String> plates = new ArrayDeque<>();
plates.push("blue");
plates.push("red");
plates.push("green");
IO.println("Top: " + plates.peek());
IO.println("Taken: " + plates.pop() + ", left: " + plates);
````

🧪 **Try it:** serve everyone in a `while (!line.isEmpty())` loop. Then call `poll()` on the empty queue (it gives `null`) and `pop()` on an empty stack (it throws).

## 6. Which one to use?

| You need… | Use |
|---|---|
| items in order, repeats allowed, access by position | `List` (`ArrayList`) |
| each item once, fast "is it there?" | `Set` (`HashSet`; `LinkedHashSet` for arrival order, `TreeSet` for sorted) |
| a value for each key | `Map` (`HashMap`; `LinkedHashMap`, `TreeMap`) |
| first in, first out, or last in, first out | `ArrayDeque` as a queue or a stack |

**Watch out:** `map.get(key)` gives `null` for a missing key, and using that `null` as a number stops the program with a `NullPointerException`. Prefer `getOrDefault`.

````java
Map<String, Integer> stock = new HashMap<>();
stock.put("pens", 12);
Integer pencils = stock.get("pencils");
IO.println(pencils);                              // null: no such key
IO.println(stock.getOrDefault("pencils", 0) + 5); // safe
````

🧪 **Try it:** write `int n = stock.get("pencils");` and run: that's the NullPointerException. Then change it to use `getOrDefault`.

## In short

Collections grow and shrink as you add and remove items. A **List** keeps the order and allows repeats; a **Set** keeps one copy of each item; a **Map** links each **key** to a value.

```java
void main() {
    Map<String, Double> prices = new HashMap<>();
    prices.put("apple", 0.5);
    prices.put("bread", 2.25);
    List<String> cart = new ArrayList<>();
    cart.add("apple");
    cart.add("bread");
    cart.add("apple");
    double total = 0;
    for (String item : cart) {
        total += prices.getOrDefault(item, 0.0);
    }
    Set<String> kinds = new HashSet<>(cart);
    IO.println(cart.size() + " items, " + kinds.size() + " different, total " + total);
}
```

`Hash…` collections are fast but unordered, `Linked…` ones keep the order items arrived in, and `Tree…` ones keep them sorted. An `ArrayDeque` works as a queue or a stack.

**Watch out:** collections hold objects, so write `List<Integer>`, not `List<int>`.
