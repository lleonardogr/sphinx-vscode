## 1. A list that grows: ArrayList

An array's size is fixed. A **List** grows and shrinks as you add and remove items. You create one with the type of its items between `< >`:

```java
List<String> names = new ArrayList<>();
```

| Method | Does |
|---|---|
| `add(x)` | puts x at the end |
| `get(i)` | the item at index i |
| `set(i, x)` | replaces the item at index i |
| `remove(i)` | removes the item at index i |
| `size()` | how many items |
| `contains(x)` | whether x is in it |

Collections hold **objects**, so numbers use the class names: `List<Integer>`, `List<Double>`, never `List<int>`.

````java
List<String> names = new ArrayList<>();
names.add("Ana");
names.add("Bia");
names.add("Caio");
IO.println(names + " has " + names.size() + " names");
IO.println("Second: " + names.get(1));
names.set(0, "Alice");
names.remove(1);
IO.println(names);
IO.println(names.contains("Bia"));
````

## 2. Going through a list

A list is gone through like an array: a for-each to read every item, or a `for` with `get(i)` when you need the index. Don't remove items inside a for-each (Java stops with a `ConcurrentModificationException`); use `removeIf` instead.

````java
List<Integer> scores = new ArrayList<>(List.of(72, 95, 40, 88, 35));
int total = 0;
for (int s : scores) {
    total += s;
}
IO.println("Average: " + (double) total / scores.size());
for (int i = 0; i < scores.size(); i++) {
    IO.println((i + 1) + "º: " + scores.get(i));
}
scores.removeIf(s -> s < 50);   // removes every score below 50
IO.println(scores);
````

🧪 **Try it:** `List.of(...)` alone makes a list that can't change: try `List.of(1, 2).add(3)` and read the exception. That's why the cell wraps it in `new ArrayList<>(...)`.

## 3. Inserting, finding and sorting

`add(i, x)` inserts at index `i` and moves the rest one place to the right; `indexOf(x)` gives the first index of `x`, or -1. `Collections.sort(list)` (or `list.sort(null)`) puts the items in their natural order, and `Collections.max` finds the largest.

````java
List<String> line = new ArrayList<>(List.of("Ana", "Caio", "Davi"));
line.add(1, "Bia");                  // squeezes in at index 1
IO.println(line);
IO.println(line.indexOf("Caio") + " " + line.indexOf("Zoe"));
List<Integer> scores = new ArrayList<>(List.of(72, 95, 60, 88));
Collections.sort(scores);
IO.println(scores + ", best " + Collections.max(scores));
````

## 4. Set: one copy of each item

A **Set** keeps each item at most once: adding a repeat does nothing, and `add` returns `false`. It's the tool for "the different ones". The order depends on the kind: `HashSet` has no useful order, `LinkedHashSet` keeps the order items arrived in, and `TreeSet` keeps them sorted. The slides compare a List, a Set and a Map getting the same items.

#### 🎞️ List, Set and Map, step by step

![The same additions in a List (keeps a, b, a), a Set (keeps a, b) and a Map (a key put twice keeps the last value)](collections-steps.svg)

````java
List<String> visits = List.of("home", "news", "home", "shop", "news", "home");
Set<String> pages = new LinkedHashSet<>(visits);
IO.println(visits.size() + " visits to " + pages.size() + " pages: " + pages);
IO.println(pages.add("home"));    // already there
IO.println(pages.add("help"));
IO.println(new TreeSet<>(pages)); // sorted
````

## 5. Set operations

Sets answer questions about groups: `addAll` gives the **union** (everything in either), `retainAll` the **intersection** (only what's in both), and `removeAll` the **difference** (what's in the first but not the second). Each one changes the set it's called on, so start from a copy.

````java
Set<String> java = Set.of("Ana", "Bia", "Caio");
Set<String> c = Set.of("Bia", "Caio", "Davi");
Set<String> both = new TreeSet<>(java);
both.retainAll(c);
Set<String> any = new TreeSet<>(java);
any.addAll(c);
Set<String> onlyJava = new TreeSet<>(java);
onlyJava.removeAll(c);
IO.println("both: " + both + ", any: " + any + ", only Java: " + onlyJava);
````

## 6. Common mistakes

- **`List<int>`:** collections hold objects; write `List<Integer>`.
- **`remove` on a `List<Integer>`:** `list.remove(1)` removes the item **at index 1**; to remove the number 1, write `list.remove(Integer.valueOf(1))`.
- **Removing inside a for-each:** Java stops with a `ConcurrentModificationException`; use `removeIf`.
- **Changing a `List.of(...)`:** it's fixed; copy it into a `new ArrayList<>(...)` first.

````java
List<Integer> numbers = new ArrayList<>(List.of(5, 1, 7));
numbers.remove(1);                     // the item at index 1, which is the number 1
IO.println(numbers);
numbers.remove(Integer.valueOf(7));    // the number 7
IO.println(numbers);
````

## In short

Collections grow and shrink as you add and remove items, unlike arrays. A **List** keeps the order and allows repeats; a **Set** keeps one copy of each item.

```java
void main() {
    List<String> cart = new ArrayList<>();
    cart.add("apple");
    cart.add("bread");
    cart.add("apple");
    cart.add(1, "milk");                 // at index 1
    IO.println(cart + " " + cart.size());
    cart.remove("bread");
    IO.println(cart.get(0) + " " + cart.contains("milk"));

Set<String> kinds = new TreeSet<>(cart);
    IO.println(kinds);                   // [apple, milk]: sorted, no repeats
}
```

`HashSet` is fastest but has no useful order, `LinkedHashSet` keeps the order items arrived in, and `TreeSet` keeps them sorted.

**Watch out:** collections hold objects, so write `List<Integer>`, not `List<int>`.
