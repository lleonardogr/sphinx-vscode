## 1. Uma lista que cresce: ArrayList

O tamanho de um array é fixo. Uma **List** cresce e diminui conforme você põe e tira itens. Você a cria com o tipo dos itens entre `< >`:

```java
List<String> names = new ArrayList<>();
```

| Método | Faz |
|---|---|
| `add(x)` | põe x no fim |
| `get(i)` | o item no índice i |
| `set(i, x)` | troca o item no índice i |
| `remove(i)` | tira o item no índice i |
| `size()` | quantos itens |
| `contains(x)` | se x está nela |

Coleções guardam **objetos**, então números usam os nomes das classes: `List<Integer>`, `List<Double>`, nunca `List<int>`.

````java
List<String> names = new ArrayList<>();
names.add("Ana");
names.add("Bia");
names.add("Caio");
IO.println(names + " tem " + names.size() + " nomes");
IO.println("Segundo: " + names.get(1));
names.set(0, "Alice");
names.remove(1);
IO.println(names);
IO.println(names.contains("Bia"));
````

🧪 **Experimente:** preveja a lista depois do `set` e do `remove` antes de rodar. Depois chame `names.get(5)`: listas conferem os índices, como arrays.

## 2. Percorrendo uma lista

Uma lista é percorrida como um array: um for-each para ler cada item, ou um `for` com `get(i)` quando precisa do índice. Não remova itens dentro de um for-each (o Java para com uma `ConcurrentModificationException`); use `removeIf` no lugar.

````java
List<Integer> scores = new ArrayList<>(List.of(72, 95, 40, 88, 35));
int total = 0;
for (int s : scores) {
    total += s;
}
IO.println("Média: " + (double) total / scores.size());
for (int i = 0; i < scores.size(); i++) {
    IO.println((i + 1) + "º: " + scores.get(i));
}
scores.removeIf(s -> s < 50);   // tira toda nota abaixo de 50
IO.println(scores);
````

🧪 **Experimente:** `List.of(...)` sozinho cria uma lista que não muda: tente `List.of(1, 2).add(3)` e leia a exceção. É por isso que a célula a embrulha em `new ArrayList<>(...)`.

## 3. Inserindo, achando e ordenando

`add(i, x)` insere no índice `i` e empurra o resto uma posição para a direita; `indexOf(x)` devolve o primeiro índice de `x`, ou -1. `Collections.sort(lista)` (ou `lista.sort(null)`) põe os itens na ordem natural, e `Collections.max` acha o maior.

````java
List<String> line = new ArrayList<>(List.of("Ana", "Caio", "Davi"));
line.add(1, "Bia");                  // entra no índice 1
IO.println(line);
IO.println(line.indexOf("Caio") + " " + line.indexOf("Zoe"));
List<Integer> scores = new ArrayList<>(List.of(72, 95, 60, 88));
Collections.sort(scores);
IO.println(scores + ", melhor " + Collections.max(scores));
````

## 4. Set: uma cópia de cada item

Um **Set** guarda cada item no máximo uma vez: adicionar um repetido não faz nada, e o `add` devolve `false`. É a ferramenta para "os diferentes". A ordem depende do tipo: `HashSet` não tem ordem útil, `LinkedHashSet` mantém a ordem de chegada, e `TreeSet` os mantém ordenados. Os slides comparam uma List, um Set e um Map recebendo os mesmos itens.

#### 🎞️ List, Set e Map, passo a passo

![As mesmas inserções numa List (guarda a, b, a), num Set (guarda a, b) e num Map (uma chave posta duas vezes fica com o último valor)](collections-steps.pt-br.svg)

````java
List<String> visits = List.of("inicio", "noticias", "inicio", "loja", "noticias", "inicio");
Set<String> pages = new LinkedHashSet<>(visits);
IO.println(visits.size() + " visitas a " + pages.size() + " páginas: " + pages);
IO.println(pages.add("inicio"));  // já está lá
IO.println(pages.add("ajuda"));
IO.println(new TreeSet<>(pages)); // ordenado
````

## 5. Operações com conjuntos

Conjuntos respondem perguntas sobre grupos: `addAll` dá a **união** (tudo o que está em algum), `retainAll` a **interseção** (só o que está nos dois), e `removeAll` a **diferença** (o que está no primeiro mas não no segundo). Cada um muda o conjunto em que é chamado, então comece de uma cópia.

````java
Set<String> java = Set.of("Ana", "Bia", "Caio");
Set<String> c = Set.of("Bia", "Caio", "Davi");
Set<String> both = new TreeSet<>(java);
both.retainAll(c);
Set<String> any = new TreeSet<>(java);
any.addAll(c);
Set<String> onlyJava = new TreeSet<>(java);
onlyJava.removeAll(c);
IO.println("os dois: " + both + ", algum: " + any + ", só Java: " + onlyJava);
````

## 6. Erros comuns

- **`List<int>`:** coleções guardam objetos; escreva `List<Integer>`.
- **`remove` numa `List<Integer>`:** `lista.remove(1)` tira o item **no índice 1**; para tirar o número 1, escreva `lista.remove(Integer.valueOf(1))`.
- **Remover dentro de um for-each:** o Java para com uma `ConcurrentModificationException`; use `removeIf`.
- **Mudar um `List.of(...)`:** ele é fixo; copie num `new ArrayList<>(...)` antes.

````java
List<Integer> numbers = new ArrayList<>(List.of(5, 1, 7));
numbers.remove(1);                     // o item no índice 1, que é o número 1
IO.println(numbers);
numbers.remove(Integer.valueOf(7));    // o número 7
IO.println(numbers);
````

## Em resumo

Coleções crescem e diminuem conforme você põe e tira itens, ao contrário dos arrays. Uma **List** mantém a ordem e aceita repetidos; um **Set** guarda uma cópia de cada item.

```java
void main() {
    List<String> cart = new ArrayList<>();
    cart.add("maçã");
    cart.add("pão");
    cart.add("maçã");
    cart.add(1, "leite");                // no índice 1
    IO.println(cart + " " + cart.size());
    cart.remove("pão");
    IO.println(cart.get(0) + " " + cart.contains("leite"));

Set<String> kinds = new TreeSet<>(cart);
    IO.println(kinds);                   // [leite, maçã]: ordenado, sem repetidos
}
```

`HashSet` é o mais rápido, mas não tem ordem útil, `LinkedHashSet` mantém a ordem de chegada, e `TreeSet` os mantém ordenados.

**Cuidado:** coleções guardam objetos, então escreva `List<Integer>`, não `List<int>`.
