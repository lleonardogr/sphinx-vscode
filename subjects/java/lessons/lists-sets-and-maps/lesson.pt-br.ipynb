## 1. Uma lista que cresce: ArrayList

O tamanho de um array é fixo. Uma **List** cresce e diminui conforme você põe e tira itens. Você a cria com o tipo dos itens entre `< >`:

```java
List<String> names = new ArrayList<>();
```

| Método | Faz |
|---|---|
| `add(x)` | põe x no fim |
| `get(i)` | o item no índice i |
| `set(i, x)` | troca o item no índice i |
| `remove(i)` | tira o item no índice i |
| `size()` | quantos itens |
| `contains(x)` | se x está nela |

Coleções guardam **objetos**, então números usam os nomes das classes: `List<Integer>`, `List<Double>`, nunca `List<int>`.

````java
List<String> names = new ArrayList<>();
names.add("Ana");
names.add("Bia");
names.add("Caio");
IO.println(names + " tem " + names.size() + " nomes");
IO.println("Segundo: " + names.get(1));
names.set(0, "Alice");
names.remove(1);
IO.println(names);
IO.println(names.contains("Bia"));
````

🧪 **Experimente:** preveja a lista depois do `set` e do `remove` antes de rodar. Depois chame `names.get(5)`: listas conferem os índices, como arrays.

## 2. Percorrendo uma lista

Uma lista é percorrida como um array: um for-each para ler cada item, ou um `for` com `get(i)` quando precisa do índice. Não remova itens dentro de um for-each (o Java para com uma `ConcurrentModificationException`); use `removeIf` no lugar.

````java
List<Integer> scores = new ArrayList<>(List.of(72, 95, 40, 88, 35));
int total = 0;
for (int s : scores) {
    total += s;
}
IO.println("Média: " + (double) total / scores.size());
for (int i = 0; i < scores.size(); i++) {
    IO.println((i + 1) + "º: " + scores.get(i));
}
scores.removeIf(s -> s < 50);   // tira toda nota abaixo de 50
IO.println(scores);
````

🧪 **Experimente:** `List.of(...)` sozinho cria uma lista que não muda: tente `List.of(1, 2).add(3)` e leia a exceção. É por isso que a célula a embrulha em `new ArrayList<>(...)`.

## 3. Set: uma cópia de cada item

Um **Set** guarda cada item no máximo uma vez: adicionar um repetido não faz nada, e o `add` devolve `false`. É a ferramenta para "os diferentes". A ordem depende do tipo: `HashSet` não tem ordem útil, `LinkedHashSet` mantém a ordem de chegada, e `TreeSet` os mantém ordenados. Os slides comparam uma List, um Set e um Map recebendo os mesmos itens.

#### 🎞️ List, Set e Map, passo a passo

![As mesmas inserções numa List (guarda a, b, a), num Set (guarda a, b) e num Map (uma chave posta duas vezes fica com o último valor)](collections-steps.pt-br.svg)

````java
List<String> visits = List.of("inicio", "noticias", "inicio", "loja", "noticias", "inicio");
Set<String> pages = new LinkedHashSet<>(visits);
IO.println(visits.size() + " visitas a " + pages.size() + " páginas: " + pages);
IO.println(pages.add("inicio"));  // já está lá
IO.println(pages.add("ajuda"));
IO.println(new TreeSet<>(pages)); // ordenado
````

🧪 **Experimente:** troque `LinkedHashSet` por `HashSet` e compare a ordem. Depois verifique se uma página foi visitada com `pages.contains(...)`.

## 4. Map: buscando pela chave

Um **Map** liga cada **chave** a um **valor**, como um dicionário liga uma palavra ao significado. `put(chave, valor)` guarda (e troca o valor antigo se a chave já existia), `get(chave)` lê (`null` quando a chave não existe), `getOrDefault(chave, x)` devolve `x` no lugar de `null`, e `containsKey` pergunta. Para percorrer, faça um laço em `entrySet()`: cada entrada tem `getKey()` e `getValue()`.

````java
Map<String, Double> prices = new HashMap<>();
prices.put("maçã", 0.5);
prices.put("pão", 2.25);
prices.put("maçã", 0.6);           // troca o 0.5
IO.println(prices.get("maçã"));
IO.println(prices.get("leite"));   // não está lá
IO.println(prices.getOrDefault("leite", 0.0));

List<String> cart = List.of("maçã", "pão", "maçã", "leite");
double total = 0;
for (String item : cart) {
    total += prices.getOrDefault(item, 0.0);
}
IO.println("Total: " + total);
for (Map.Entry<String, Double> e : prices.entrySet()) {
    IO.println(e.getKey() + " custa " + e.getValue());
}
````

## 5. Filas e pilhas: ArrayDeque

Um `ArrayDeque` pode ser usado pelas duas pontas. Como **fila** (o primeiro a entrar é o primeiro a sair, como numa fila de banco): `offer` entra no fim, `poll` tira da frente. Como **pilha** (o último a entrar é o primeiro a sair, como uma pilha de pratos): `push` põe no topo, `pop` tira do topo. `peek` olha sem tirar.

````java
Deque<String> line = new ArrayDeque<>();
line.offer("Ana");
line.offer("Bia");
line.offer("Caio");
IO.println("Atendida: " + line.poll() + ", esperando: " + line);

Deque<String> plates = new ArrayDeque<>();
plates.push("azul");
plates.push("vermelho");
plates.push("verde");
IO.println("Topo: " + plates.peek());
IO.println("Tirado: " + plates.pop() + ", sobraram: " + plates);
````

🧪 **Experimente:** atenda todo mundo num laço `while (!line.isEmpty())`. Depois chame `poll()` na fila vazia (devolve `null`) e `pop()` numa pilha vazia (lança uma exceção).

## 6. Qual usar?

| Você precisa de… | Use |
|---|---|
| itens em ordem, com repetidos, acesso pela posição | `List` (`ArrayList`) |
| cada item uma vez, "está lá?" rápido | `Set` (`HashSet`; `LinkedHashSet` para a ordem de chegada, `TreeSet` para ordenado) |
| um valor para cada chave | `Map` (`HashMap`; `LinkedHashMap`, `TreeMap`) |
| o primeiro a entrar sai primeiro, ou o último sai primeiro | `ArrayDeque` como fila ou pilha |

**Cuidado:** `map.get(chave)` devolve `null` para uma chave que não existe, e usar esse `null` como número para o programa com uma `NullPointerException`. Prefira `getOrDefault`.

````java
Map<String, Integer> stock = new HashMap<>();
stock.put("canetas", 12);
Integer pencils = stock.get("lapis");
IO.println(pencils);                              // null: chave inexistente
IO.println(stock.getOrDefault("lapis", 0) + 5);   // seguro
````

🧪 **Experimente:** escreva `int n = stock.get("lapis");` e rode: essa é a NullPointerException. Depois mude para usar `getOrDefault`.

## Em resumo

Coleções crescem e encolhem conforme você adiciona e remove itens. Uma **List** mantém a ordem e aceita repetidos; um **Set** guarda uma cópia de cada item; um **Map** liga cada **chave** a um valor.

```java
void main() {
    Map<String, Double> prices = new HashMap<>();
    prices.put("maçã", 0.5);
    prices.put("pão", 2.25);
    List<String> cart = new ArrayList<>();
    cart.add("maçã");
    cart.add("pão");
    cart.add("maçã");
    double total = 0;
    for (String item : cart) {
        total += prices.getOrDefault(item, 0.0);
    }
    Set<String> kinds = new HashSet<>(cart);
    IO.println(cart.size() + " itens, " + kinds.size() + " diferentes, total " + total);
}
```

Coleções `Hash…` são rápidas mas sem ordem, as `Linked…` mantêm a ordem de chegada e as `Tree…` mantêm tudo ordenado. Um `ArrayDeque` funciona como fila ou pilha.

**Cuidado:** coleções guardam objetos, então escreva `List<Integer>`, não `List<int>`.
