## Em resumo

Coleções crescem e encolhem conforme você adiciona e remove itens. Uma **List** mantém a ordem e aceita repetidos; um **Set** guarda uma cópia de cada item; um **Map** liga cada **chave** a um valor.

#### 🎞️ List, Set e Map, passo a passo

![As mesmas inserções numa List (guarda a, b, a), num Set (guarda a, b) e num Map (uma chave posta duas vezes fica com o último valor)](collections-steps.pt-br.svg)

```java
void main() {
    Map<String, Double> prices = new HashMap<>();
    prices.put("maçã", 0.5);
    prices.put("pão", 2.25);
    List<String> cart = new ArrayList<>();
    cart.add("maçã");
    cart.add("pão");
    cart.add("maçã");
    double total = 0;
    for (String item : cart) {
        total += prices.getOrDefault(item, 0.0);
    }
    Set<String> kinds = new HashSet<>(cart);
    IO.println(cart.size() + " itens, " + kinds.size() + " diferentes, total " + total);
}
```

Coleções `Hash…` são rápidas mas sem ordem, as `Linked…` mantêm a ordem de chegada e as `Tree…` mantêm tudo ordenado. Um `ArrayDeque` funciona como fila ou pilha.

**As mesmas palavras em quatro coleções:** repare quais mantêm a ordem, quais descartam repetidas e quais ordenam.

````java
List<String> words = List.of("pear", "apple", "fig", "apple", "kiwi");
IO.println("ArrayList:     " + new ArrayList<>(words));
IO.println("HashSet:       " + new HashSet<>(words));
IO.println("LinkedHashSet: " + new LinkedHashSet<>(words));
IO.println("TreeSet:       " + new TreeSet<>(words));

List<Integer> numbers = new ArrayList<>(List.of(10, 20, 30));
numbers.remove(1);                    // remove o item no índice 1
numbers.remove(Integer.valueOf(10));  // remove o valor 10
IO.println(numbers);
````
