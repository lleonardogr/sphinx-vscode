## 1. A String is a row of numbered characters

A `String` is a sequence of characters, and each one has a position, its **index**, starting at **0**. In `"java"`, `'j'` is at 0 and `'a'` at 3. `length()` counts the characters, and `charAt(i)` gives the one at index `i`. The last index is always `length() - 1`.

````java
String word = "java";
IO.println(word.length());
IO.println(word.charAt(0));
IO.println(word.charAt(3));
IO.println(word.charAt(word.length() - 1));   // always the last one
````

🧪 **Try it:** try `word.charAt(4)` and read the exception: index 4 doesn't exist in a 4-letter word. Then try an empty String `""`: what is its length?

## 2. char: a single character

`charAt` gives a `char`, written in **single** quotes: `'a'`. A `char` is really a small number (its code), so you compare chars with `==` and can even do arithmetic: `'a' + 1` is the code of `'b'`. The `Character` class answers questions about one: `isDigit`, `isLetter`, `isUpperCase`, `toUpperCase`…

````java
char c = 'b';
IO.println(c == 'b');
IO.println((int) c);              // its code
IO.println((char) (c + 1));       // the next letter
IO.println(Character.isDigit('7') + " " + Character.isLetter('7'));
IO.println(Character.toUpperCase(c));
````

🧪 **Try it:** try `c = 'Z'` and `c = '?'`. Is `'a' == "a"` allowed? Try it: a char and a String are different types.

## 3. Going through a String with a loop

To look at every character, loop over the indexes from 0 to `length() - 1` and read each one with `charAt(i)`. This pattern solves a huge number of text problems.

````java
String text = "Hi 2 you";
int digits = 0;
for (int i = 0; i < text.length(); i++) {
    char c = text.charAt(i);
    IO.println(i + ": '" + c + "'");
    if (Character.isDigit(c)) {
        digits++;
    }
}
IO.println("Digits: " + digits);
````

## 4. Finding: indexOf, contains, startsWith

`indexOf(x)` gives the index where `x` first appears, or **-1** when it isn't there; `lastIndexOf` searches from the end. `contains`, `startsWith` and `endsWith` answer with `true` or `false`.

````java
String email = "ada.lovelace@example.org";
IO.println(email.indexOf('@'));
IO.println(email.indexOf('.'));
IO.println(email.lastIndexOf('.'));
IO.println(email.indexOf('#'));          // not there
IO.println(email.contains("love"));
IO.println(email.endsWith(".org"));
````

🧪 **Try it:** count the characters by hand to check the first answer. Then search for `"example"`: what index does a word give?

## 5. Cutting: substring

`substring(start, end)` copies the characters from `start` up to, **but not including**, `end`. So the piece has `end - start` characters. `substring(start)` copies from `start` to the end. Combined with `indexOf`, it cuts text at a separator.

````java
String email = "ada@example.org";
int at = email.indexOf('@');
String user = email.substring(0, at);       // before the @
String domain = email.substring(at + 1);    // after the @
IO.println(user + " | " + domain);
IO.println("java".substring(1, 3));
````

## 6. A String never changes

No String method changes the String itself. `toUpperCase()`, `trim()` or `replace(...)` **return a new String**, and the original stays as it was. To keep the result, store it: `name = name.trim();`. The slides show the new objects appearing in the heap.

#### 🎞️ Strings never change

![toUpperCase and + create new String objects in the heap; the variables are pointed at the new ones and the old text is left behind](string-memory-steps.svg)

````java
String s = "java";
s.toUpperCase();                 // result thrown away
IO.println(s);
String loud = s.toUpperCase();   // stored in another variable
IO.println(s + " " + loud);
s = s + "!";                     // s now points to a new String
IO.println(s);
````

## 7. Comparing text

Compare the letters of two Strings with `equals`, never `==` (which asks whether they're the same object). `equalsIgnoreCase` ignores upper and lower case. `a.compareTo(b)` puts them in dictionary order: negative when `a` comes first, 0 when equal, positive when `a` comes after.

````java
String a = "apple";
String b = "Apple";
IO.println(a.equals(b));
IO.println(a.equalsIgnoreCase(b));
IO.println("apple".compareTo("banana") < 0);   // apple comes first
IO.println("pear".compareTo("peach") > 0);
````

🧪 **Try it:** compare `"Zebra"` with `"apple"` using `compareTo`: which comes first? Upper-case letters come before all lower-case ones in their codes.

## 8. Handy methods: trim, split, repeat, replace

| Method | Does |
|---|---|
| `trim()` | removes spaces at both ends |
| `split(" ")` | cuts at each separator, giving an array of pieces |
| `String.join("-", parts)` | joins pieces with a separator |
| `repeat(n)` | the text n times |
| `replace(a, b)` | swaps every `a` for `b` |
| `toLowerCase()` / `toUpperCase()` | changes the case |

````java
String line = "  red green blue  ";
String clean = line.trim();
String[] colors = clean.split(" ");
IO.println(colors.length + " colors, the first is " + colors[0]);
IO.println(String.join(" / ", colors));
IO.println("ab".repeat(3));
IO.println(clean.replace("green", "yellow"));
````

🧪 **Try it:** split `line` without `trim()` first: how many pieces now, and what are the extra ones? Then split `"a,b,,c"` at `","`.

## 9. Building text with StringBuilder

Since a String never changes, `text = text + piece` in a loop creates a new String every round. A `StringBuilder` is a String you **can** change: `append` adds to the end, and `toString()` gives the final String. Use it when you build text piece by piece.

````java
StringBuilder sb = new StringBuilder();
for (int i = 1; i <= 5; i++) {
    if (i > 1) {
        sb.append(", ");
    }
    sb.append(i);
}
String result = sb.toString();
IO.println(result);
````

## In short

A **String** is a sequence of characters, numbered from 0. `charAt(i)` gives one `char`, `length()` counts them, `indexOf` finds one, and `substring(start, end)` copies a piece, from `start` up to but **not including** `end`.

```java
void main() {
    String email = IO.readln().trim();
    int at = email.indexOf('@');
    String user = email.substring(0, at);
    String domain = email.substring(at + 1);
    IO.println("User: " + user + " (" + user.length() + " characters)");
    int dots = 0;
    for (int i = 0; i < email.length(); i++) {
        if (email.charAt(i) == '.') {
            dots++;
        }
    }
    IO.println("Dots: " + dots);
    IO.println(domain.toLowerCase().equals("gmail.com") ? "Gmail" : "Other");
}
```

A String never changes: `toUpperCase()` returns a new one. To build text in a loop, append to a `StringBuilder`.

**Watch out:** compare texts with `a.equals(b)`, never `a == b`. Single characters (`char`, in single quotes) do use `==`.
