## 1. Uma String é uma fileira de caracteres numerados

Uma `String` é uma sequência de caracteres, e cada um tem uma posição, o **índice**, começando em **0**. Em `"java"`, `'j'` está em 0 e `'a'` em 3. `length()` conta os caracteres, e `charAt(i)` devolve o que está no índice `i`. O último índice é sempre `length() - 1`.

````java
String word = "java";
IO.println(word.length());
IO.println(word.charAt(0));
IO.println(word.charAt(3));
IO.println(word.charAt(word.length() - 1));   // sempre o último
````

## 2. char: um caractere só

O `charAt` devolve um `char`, escrito entre aspas **simples**: `'a'`. Um `char` é na verdade um número pequeno (o código dele), então você compara chars com `==` e pode até fazer contas: `'a' + 1` é o código de `'b'`. A classe `Character` responde perguntas sobre um char: `isDigit`, `isLetter`, `isUpperCase`, `toUpperCase`…

````java
char c = 'b';
IO.println(c == 'b');
IO.println((int) c);              // o código dele
IO.println((char) (c + 1));       // a próxima letra
IO.println(Character.isDigit('7') + " " + Character.isLetter('7'));
IO.println(Character.toUpperCase(c));
````

🧪 **Experimente:** tente `c = 'Z'` e `c = '?'`. Dá para fazer `'a' == "a"`? Tente: um char e uma String são tipos diferentes.

## 3. Percorrendo uma String com um laço

Para olhar cada caractere, percorra os índices de 0 até `length() - 1` e leia cada um com `charAt(i)`. Esse padrão resolve uma quantidade enorme de problemas com texto.

````java
String text = "Oi 2 você";
int digits = 0;
for (int i = 0; i < text.length(); i++) {
    char c = text.charAt(i);
    IO.println(i + ": '" + c + "'");
    if (Character.isDigit(c)) {
        digits++;
    }
}
IO.println("Dígitos: " + digits);
````

## 4. Procurando: indexOf, contains, startsWith

`indexOf(x)` devolve o índice onde `x` aparece pela primeira vez, ou **-1** quando não aparece; `lastIndexOf` procura a partir do fim. `contains`, `startsWith` e `endsWith` respondem com `true` ou `false`.

````java
String email = "ada.lovelace@example.org";
IO.println(email.indexOf('@'));
IO.println(email.indexOf('.'));
IO.println(email.lastIndexOf('.'));
IO.println(email.indexOf('#'));          // não está lá
IO.println(email.contains("love"));
IO.println(email.endsWith(".org"));
````

## 5. Recortando: substring

`substring(início, fim)` copia os caracteres de `início` até, **sem incluir**, `fim`. Então o pedaço tem `fim - início` caracteres. `substring(início)` copia de `início` até o fim. Junto com o `indexOf`, ele corta um texto num separador.

````java
String email = "ada@example.org";
int at = email.indexOf('@');
String user = email.substring(0, at);       // antes do @
String domain = email.substring(at + 1);    // depois do @
IO.println(user + " | " + domain);
IO.println("java".substring(1, 3));
````

## 6. Uma String nunca muda

Nenhum método de String muda a própria String. `toUpperCase()`, `trim()` ou `replace(...)` **devolvem uma String nova**, e a original continua como estava. Para guardar o resultado, atribua: `name = name.trim();`. Os slides mostram os objetos novos aparecendo no heap.

#### 🎞️ Strings nunca mudam

![toUpperCase e + criam novos objetos String no heap; as variáveis passam a apontar para os novos e o texto antigo fica para trás](string-memory-steps.pt-br.svg)

````java
String s = "java";
s.toUpperCase();                 // resultado jogado fora
IO.println(s);
String loud = s.toUpperCase();   // guardado em outra variável
IO.println(s + " " + loud);
s = s + "!";                     // s agora aponta para uma String nova
IO.println(s);
````

## 7. Comparando textos

Compare as letras de duas Strings com `equals`, nunca `==` (que pergunta se são o mesmo objeto). `equalsIgnoreCase` ignora maiúsculas e minúsculas. `a.compareTo(b)` as coloca em ordem de dicionário: negativo quando `a` vem antes, 0 quando são iguais, positivo quando `a` vem depois.

````java
String a = "maçã";
String b = "Maçã";
IO.println(a.equals(b));
IO.println(a.equalsIgnoreCase(b));
IO.println("abacaxi".compareTo("banana") < 0);   // abacaxi vem antes
IO.println("pera".compareTo("pequi") > 0);
````

## 8. Métodos úteis: trim, split, repeat, replace

| Método | Faz |
|---|---|
| `trim()` | tira os espaços das duas pontas |
| `split(" ")` | corta em cada separador, devolvendo um array de pedaços |
| `String.join("-", partes)` | junta pedaços com um separador |
| `repeat(n)` | o texto n vezes |
| `replace(a, b)` | troca cada `a` por `b` |
| `toLowerCase()` / `toUpperCase()` | muda a caixa |

````java
String line = "  vermelho verde azul  ";
String clean = line.trim();
String[] colors = clean.split(" ");
IO.println(colors.length + " cores, a primeira é " + colors[0]);
IO.println(String.join(" / ", colors));
IO.println("ab".repeat(3));
IO.println(clean.replace("verde", "amarelo"));
````

## 9. Montando texto com StringBuilder

Como uma String nunca muda, `texto = texto + pedaço` num laço cria uma String nova a cada volta. Um `StringBuilder` é uma String que **pode** mudar: `append` acrescenta no fim, e `toString()` devolve a String final. Use-o quando montar um texto pedaço por pedaço.

````java
StringBuilder sb = new StringBuilder();
for (int i = 1; i <= 5; i++) {
    if (i > 1) {
        sb.append(", ");
    }
    sb.append(i);
}
String result = sb.toString();
IO.println(result);
````

## Em resumo

Uma **String** é uma sequência de caracteres, numerados a partir de 0. `charAt(i)` dá um `char`, `length()` conta quantos são, `indexOf` acha um, e `substring(início, fim)` copia um pedaço, de `início` até `fim`, **sem incluir** o `fim`.

```java
void main() {
    String email = IO.readln().trim();
    int at = email.indexOf('@');
    String user = email.substring(0, at);
    String domain = email.substring(at + 1);
    IO.println("Usuário: " + user + " (" + user.length() + " caracteres)");
    int dots = 0;
    for (int i = 0; i < email.length(); i++) {
        if (email.charAt(i) == '.') {
            dots++;
        }
    }
    IO.println("Pontos: " + dots);
    IO.println(domain.toLowerCase().equals("gmail.com") ? "Gmail" : "Outro");
}
```

Uma String nunca muda: `toUpperCase()` devolve uma nova. Para montar texto num laço, acrescente a um `StringBuilder`.

**Cuidado:** compare textos com `a.equals(b)`, nunca `a == b`. Caracteres sozinhos (`char`, entre aspas simples) usam `==` sim.
