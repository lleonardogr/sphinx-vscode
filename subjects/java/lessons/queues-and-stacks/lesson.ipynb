## 1. Queues and stacks with ArrayDeque

An `ArrayDeque` can be used from both ends. As a **queue** (first in, first out, like a line at the bank): `offer` joins at the back, `poll` takes from the front. As a **stack** (last in, first out, like a pile of plates): `push` puts on top, `pop` takes from the top. `peek` looks without removing.

#### 🎞️ A stack and a queue, step by step

![A stack: push A, push B, then pop takes B. A queue: D joins at the back, then poll takes A from the front.](stack-and-queue-steps.svg)

````java
Deque<String> line = new ArrayDeque<>();
line.offer("Ana");
line.offer("Bia");
line.offer("Caio");
IO.println("Served: " + line.poll() + ", waiting: " + line);

Deque<String> plates = new ArrayDeque<>();
plates.push("blue");
plates.push("red");
plates.push("green");
IO.println("Top: " + plates.peek());
IO.println("Taken: " + plates.pop() + ", left: " + plates);
````

🧪 **Try it:** serve everyone in a `while (!line.isEmpty())` loop. Then call `poll()` on the empty queue (it gives `null`) and `pop()` on an empty stack (it throws).

## 2. A queue in action: taking turns

A queue models anything served in arrival order: a printer, a help desk, players taking turns. Taking a turn is `poll()` the front and, if the player plays again later, `offer()` them back at the end.

````java
Deque<String> players = new ArrayDeque<>(List.of("Ana", "Bia", "Caio"));
for (int turn = 1; turn <= 5; turn++) {
    String now = players.poll();
    IO.println("turn " + turn + ": " + now);
    players.offer(now);                 // back to the end of the line
}
````

🧪 **Try it:** stop putting the player back (remove `players.offer(now)`) and run: what happens on turn 4?

## 3. A stack in action: undo

A stack models anything undone in reverse order: an editor's undo, a browser's Back button, nested calls. Doing something is `push`; undoing is `pop`, which always takes the most recent action.

````java
Deque<String> history = new ArrayDeque<>();
history.push("home");
history.push("news");
history.push("article");
IO.println("now on " + history.peek());
history.pop();                                  // the Back button
IO.println("back to " + history.peek());
IO.println("history, top first: " + history);
````

🧪 **Try it:** press Back three times. What does `peek()` give when the history is empty? And `pop()`?

String typed(List<String> actions) {
    Deque<String> stack = new ArrayDeque<>();
    for (String a : actions) {
        if (!a.equals("undo")) {

        } else if (!stack.isEmpty()) {

## 4. Empty queues and stacks

Before taking an item, ask `isEmpty()`. On an empty deque, `poll()` and `peek()` return `null`, while `pop()` and `remove()` throw a `NoSuchElementException`. A `while (!queue.isEmpty())` loop serves everyone and stops by itself.

````java
Deque<String> line = new ArrayDeque<>(List.of("Ana", "Bia"));
while (!line.isEmpty()) {
    IO.println("serving " + line.poll());
}
IO.println("poll on empty: " + line.poll());
try {
    line.pop();
} catch (NoSuchElementException e) {
    IO.println("pop on empty throws " + e.getClass().getSimpleName());
}
````

🧪 **Try it:** try `line.offer(null)` and read the exception: an ArrayDeque refuses `null`, which is why `null` can safely mean "empty".

## 5. PriorityQueue: the most urgent first

A **PriorityQueue** doesn't follow arrival order: `poll()` always takes the **smallest** item (by natural order, or by a comparator you give it). It fits "the most urgent first": tasks by deadline, patients by severity.

````java
record Task(String name, int priority) {}

void main() {
    Queue<Task> todo = new PriorityQueue<>(Comparator.comparingInt(Task::priority));
    todo.offer(new Task("write report", 3));
    todo.offer(new Task("fix the server", 1));
    todo.offer(new Task("reply email", 2));
    while (!todo.isEmpty()) {
        IO.println(todo.poll());
    }
}
````

🧪 **Try it:** print `todo` itself before the loop: its order looks scrambled. Only `poll()` is guaranteed to give the smallest.

## 6. Common mistakes

- **The old `Stack` class:** it still works, but `ArrayDeque` is faster and is what the Java documentation recommends.
- **Mixing ends:** `push`/`pop` work at the front, and so do `poll`/`peek`; `offer` adds at the back. Using `push` to join a queue cuts the line.
- **`remove(0)` on an ArrayList as a queue:** it moves every other item, so it gets slow on long lines; use an ArrayDeque.
- **Iterating a PriorityQueue** expecting sorted order (step 5).

````java
Deque<String> line = new ArrayDeque<>();
line.offer("Ana");
line.offer("Bia");
line.push("Zoe");                   // push adds at the FRONT: Zoe cut the line
IO.println(line + ", next: " + line.peek());
````

🧪 **Try it:** replace `push` with `offer` so Zoe waits her turn.

## In short

A **queue** serves items in the order they arrived: first in, first out, like a line at the bank. A **stack** gives back the last item put in: last in, first out, like a pile of plates. `ArrayDeque` does both.

```java
void main() {
    Deque<String> line = new ArrayDeque<>();
    line.offer("Ana");                    // joins at the back
    line.offer("Bia");
    IO.println(line.poll() + " served, " + line + " waiting");

Deque<String> undo = new ArrayDeque<>();
    undo.push("type a");                  // on top
    undo.push("type b");
    IO.println("undo " + undo.pop());     // the last action first

Queue<Integer> urgent = new PriorityQueue<>(List.of(5, 1, 3));
    IO.println(urgent.poll());            // 1: the smallest comes out first
}
```

`peek` looks at the next item without taking it.

**Watch out:** on an empty deque, `poll` and `peek` return `null`, but `pop` throws an exception. Check `isEmpty()` first.
