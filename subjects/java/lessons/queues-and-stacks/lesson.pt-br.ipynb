## 1. Filas e pilhas com ArrayDeque

Um `ArrayDeque` pode ser usado pelas duas pontas. Como **fila** (o primeiro a entrar é o primeiro a sair, como numa fila de banco): `offer` entra no fim, `poll` tira da frente. Como **pilha** (o último a entrar é o primeiro a sair, como uma pilha de pratos): `push` põe no topo, `pop` tira do topo. `peek` olha sem tirar.

#### 🎞️ Uma pilha e uma fila, passo a passo

![Uma pilha: push A, push B, e o pop tira B. Uma fila: D entra no fundo, e o poll tira A da frente.](stack-and-queue-steps.pt-br.svg)

````java
Deque<String> line = new ArrayDeque<>();
line.offer("Ana");
line.offer("Bia");
line.offer("Caio");
IO.println("Atendida: " + line.poll() + ", esperando: " + line);

Deque<String> plates = new ArrayDeque<>();
plates.push("azul");
plates.push("vermelho");
plates.push("verde");
IO.println("Topo: " + plates.peek());
IO.println("Tirado: " + plates.pop() + ", sobraram: " + plates);
````

## 2. Uma fila em ação: revezando

Uma fila modela tudo o que é atendido em ordem de chegada: uma impressora, um balcão, jogadores revezando. Fazer uma jogada é `poll()` da frente e, se o jogador joga de novo depois, `offer()` dele de volta no fim.

````java
Deque<String> players = new ArrayDeque<>(List.of("Ana", "Bia", "Caio"));
for (int turn = 1; turn <= 5; turn++) {
    String now = players.poll();
    IO.println("rodada " + turn + ": " + now);
    players.offer(now);                 // volta para o fim da fila
}
````

## 3. Uma pilha em ação: desfazer

Uma pilha modela tudo o que se desfaz em ordem inversa: o desfazer de um editor, o Voltar do navegador, chamadas aninhadas. Fazer algo é `push`; desfazer é `pop`, que sempre pega a ação mais recente.

````java
Deque<String> history = new ArrayDeque<>();
history.push("inicio");
history.push("noticias");
history.push("artigo");
IO.println("agora em " + history.peek());
history.pop();                                  // o botão Voltar
IO.println("de volta a " + history.peek());
IO.println("histórico, topo primeiro: " + history);
````

## 4. Filas e pilhas vazias

Antes de tirar um item, pergunte `isEmpty()`. Num deque vazio, `poll()` e `peek()` devolvem `null`, enquanto `pop()` e `remove()` lançam uma `NoSuchElementException`. Um laço `while (!fila.isEmpty())` atende todo mundo e para sozinho.

````java
Deque<String> line = new ArrayDeque<>(List.of("Ana", "Bia"));
while (!line.isEmpty()) {
    IO.println("atendendo " + line.poll());
}
IO.println("poll no vazio: " + line.poll());
try {
    line.pop();
} catch (NoSuchElementException e) {
    IO.println("pop no vazio lança " + e.getClass().getSimpleName());
}
````

🧪 **Experimente:** tente `line.offer(null)` e leia a exceção: um ArrayDeque recusa `null`, e é por isso que `null` pode significar "vazio" com segurança.

## 5. PriorityQueue: o mais urgente primeiro

Uma **PriorityQueue** não segue a ordem de chegada: `poll()` sempre pega o **menor** item (pela ordem natural, ou por um comparador que você der). Ela combina com "o mais urgente primeiro": tarefas pelo prazo, pacientes pela gravidade.

````java
record Task(String name, int priority) {}

void main() {
    Queue<Task> todo = new PriorityQueue<>(Comparator.comparingInt(Task::priority));
    todo.offer(new Task("escrever relatório", 3));
    todo.offer(new Task("consertar o servidor", 1));
    todo.offer(new Task("responder email", 2));
    while (!todo.isEmpty()) {
        IO.println(todo.poll());
    }
}
````

🧪 **Experimente:** imprima o próprio `todo` antes do laço: a ordem parece bagunçada. Só o `poll()` garante devolver o menor.

## 6. Erros comuns

- **A antiga classe `Stack`:** ela ainda funciona, mas o `ArrayDeque` é mais rápido e é o que a documentação do Java recomenda.
- **Misturar as pontas:** `push`/`pop` trabalham na frente, assim como `poll`/`peek`; `offer` acrescenta no fim. Usar `push` para entrar numa fila é furar a fila.
- **`remove(0)` num ArrayList como fila:** ele move todos os outros itens, então fica lento em filas longas; use um ArrayDeque.
- **Percorrer uma PriorityQueue** esperando a ordem certa (passo 5).

````java
Deque<String> line = new ArrayDeque<>();
line.offer("Ana");
line.offer("Bia");
line.push("Zoe");                   // push põe na FRENTE: a Zoe furou a fila
IO.println(line + ", próxima: " + line.peek());
````

🧪 **Experimente:** troque o `push` por `offer` para a Zoe esperar a vez dela.

## Em resumo

Uma **fila** atende os itens na ordem em que chegaram: o primeiro a entrar é o primeiro a sair, como numa fila de banco. Uma **pilha** devolve o último item posto: o último a entrar é o primeiro a sair, como uma pilha de pratos. O `ArrayDeque` faz as duas coisas.

```java
void main() {
    Deque<String> line = new ArrayDeque<>();
    line.offer("Ana");                    // entra no fim
    line.offer("Bia");
    IO.println(line.poll() + " atendida, " + line + " esperando");

Deque<String> undo = new ArrayDeque<>();
    undo.push("digitar a");               // no topo
    undo.push("digitar b");
    IO.println("desfazer " + undo.pop()); // a última ação primeiro

Queue<Integer> urgent = new PriorityQueue<>(List.of(5, 1, 3));
    IO.println(urgent.poll());            // 1: o menor sai primeiro
}
```

`peek` olha o próximo item sem tirá-lo.

**Cuidado:** num deque vazio, `poll` e `peek` devolvem `null`, mas `pop` lança uma exceção. Confira `isEmpty()` antes.
