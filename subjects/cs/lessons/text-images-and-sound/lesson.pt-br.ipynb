## Em resumo

Os computadores guardam texto como números: cada caractere tem um **código**. No **ASCII**, `A` é 65, `a` é 97, o dígito `0` é 48 e o espaço é 32, e cada letra minúscula fica exatamente 32 depois da sua maiúscula. No Java um `char` é um número, então `'a' - 'A'` é 32.

**Caracteres são números:**

````java
IO.println((int) 'A');
IO.println((int) 'a');
IO.println('a' - 'A');
IO.println((char) ('A' + 2));
IO.println((int) '0' + " " + (int) ' ');
````

O ASCII só tem 128 códigos. O **Unicode** numera todos os caracteres de todos os idiomas, de é a 😀, com um **code point** como `U+00E9`. O **UTF-8** guarda code points em 1 a 4 bytes, mais para code points maiores: `A` ocupa 1, `é` 2, `€` e `☕` 3, e `😀` 4. Texto embaralhado como "Ã©" no lugar de "é" quer dizer que bytes UTF-8 foram lidos com a tabela errada.

![Como é, U+00E9, vira os bytes UTF-8 C3 A9](utf8.pt-br.svg)

#### 🎞️ é em UTF-8, passo a passo

![Codificando é, code point U+00E9, em UTF-8: os 11 bits preenchem o padrão 110xxxxx 10xxxxxx, dando os bytes C3 A9](utf8-steps.pt-br.svg)

**Os bytes UTF-8 de alguns caracteres.** `getBytes` vale aqui; o desafio de UTF-8 pede que você os calcule sozinho.

````java
for (String s : new String[] {"A", "é", "€", "😀"}) {
    byte[] bytes = s.getBytes(StandardCharsets.UTF_8);
    IO.println(String.format("U+%04X  %d byte(s): %s", s.codePointAt(0), bytes.length, HexFormat.ofDelimiter(" ").formatHex(bytes).toUpperCase()));
}
````

Imagens são grades de **pixels**, e cada pixel normalmente ocupa 3 bytes: vermelho, verde e azul, de 0 a 255, que é o que uma cor em hex como `#FF8800` escreve. O som é guardado como **amostras**: os CDs medem a onda sonora 44.100 vezes por segundo, com 16 bits por amostra e 2 canais.

**Qual o tamanho de mídia sem compressão?**

````java
long foto = 1920L * 1080 * 3; // pixels × 3 bytes (vermelho, verde, azul)
long musica = 44_100L * 2 * 2 * 180; // amostras por segundo × 2 bytes × 2 canais × 3 minutos
IO.println(String.format("uma foto Full HD: %,d bytes (%.1f MB)", foto, foto / 1e6));
IO.println(String.format("uma música de CD de 3 minutos: %,d bytes (%.1f MB)", musica, musica / 1e6));
````

Mídia sem compressão é grande, então ela é **comprimida**. A compressão **sem perda** (PNG, ZIP, codificação por comprimento de sequência) devolve cada bit; a compressão **com perda** (JPEG, MP3) também descarta detalhes que as pessoas mal percebem e deixa os arquivos dez vezes menores.
