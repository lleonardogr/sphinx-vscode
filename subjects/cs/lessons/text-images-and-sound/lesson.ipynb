## In short

Computers store text as numbers: each character has a **code**. In **ASCII**, `A` is 65, `a` is 97, the digit `0` is 48 and the space is 32, and each lower-case letter is exactly 32 after its upper-case one. In Java a `char` is a number, so `'a' - 'A'` is 32.

**Characters are numbers:**

```java
IO.println((int) 'A');
IO.println((int) 'a');
IO.println('a' - 'A');
IO.println((char) ('A' + 2));
IO.println((int) '0' + " " + (int) ' ');
```

ASCII has only 128 codes. **Unicode** numbers every character of every language, from é to 😀, with a **code point** such as `U+00E9`. **UTF-8** stores code points in 1 to 4 bytes, more for bigger code points: `A` takes 1, `é` 2, `€` and `☕` 3, and `😀` 4. Garbled text like "Ã©" instead of "é" means UTF-8 bytes were read with the wrong table.

![How é, U+00E9, becomes the UTF-8 bytes C3 A9](utf8.svg)

**The UTF-8 bytes of a few characters.** `getBytes` is fine here; the UTF-8 challenge asks you to work them out yourself.

```java
for (String s : new String[] {"A", "é", "€", "😀"}) {
    byte[] bytes = s.getBytes(StandardCharsets.UTF_8);
    IO.println(String.format("U+%04X  %d byte(s): %s", s.codePointAt(0), bytes.length, HexFormat.ofDelimiter(" ").formatHex(bytes).toUpperCase()));
}
```

Images are grids of **pixels**, and each pixel usually takes 3 bytes: red, green and blue, from 0 to 255, which is what a hex color like `#FF8800` writes. Sound is stored as **samples**: CDs measure the sound wave 44,100 times per second, with 16 bits per sample and 2 channels.

**How big is raw media?**

```java
long photo = 1920L * 1080 * 3; // pixels × 3 bytes (red, green, blue)
long song = 44_100L * 2 * 2 * 180; // samples per second × 2 bytes × 2 channels × 3 minutes
IO.println(String.format("a Full HD photo: %,d bytes (%.1f MB)", photo, photo / 1e6));
IO.println(String.format("a 3-minute CD song: %,d bytes (%.1f MB)", song, song / 1e6));
```

Raw media is big, so it is **compressed**. **Lossless** compression (PNG, ZIP, run-length encoding) gives back every bit; **lossy** compression (JPEG, MP3) also drops details people barely notice and gets files ten times smaller.
