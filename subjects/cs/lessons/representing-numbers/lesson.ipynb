## In short

Bits have no minus sign, so computers use **two's complement**: in a byte, the leftmost bit is worth **−128** instead of +128, and the others keep their values. A byte then holds −128 to 127, and ordinary addition still works.

![The byte 11111011 in two's complement is −5](signed-byte.svg)

To negate a number, invert every bit and add 1: 5 is `00000101`, so −5 is `11111010` + 1 = `11111011`. Java's `int` uses 32 bits, from −2,147,483,648 to 2,147,483,647. When a result doesn't fit, the extra bits are dropped and it **wraps around**: `Integer.MAX_VALUE + 1` is `-2147483648`, and Java gives no error. This is **overflow**.

**Predict, then run:** two's complement and wrap-around.

```java
IO.println(Integer.toBinaryString(-5));     // 32 bits of -5
IO.println((byte) 200);                     // 200 doesn't fit in a byte
IO.println(Integer.MAX_VALUE + 1);          // wraps around
IO.println(Math.abs(Integer.MIN_VALUE));    // a surprise
```

Fractions are harder. After the binary point the places are worth ½, ¼, ⅛, … so 0.5 and 0.25 are exact, but **0.1 repeats forever** in binary, like 1/3 in decimal. A `double` stores the closest value it can, in 64 bits: a sign, an exponent and a 52-bit fraction, which with a hidden leading 1 gives 53 bits of precision, about 15–16 decimal digits. That's why `0.1 + 0.2` prints `0.30000000000000004`.

So: compare doubles with a tolerance, `Math.abs(a - b) < 1e-9`, and keep money in whole cents in a `long`, or in `BigDecimal`.

**0.1 isn't exact in binary:**

```java
double sum = 0.1 + 0.2;
IO.println(sum);
IO.println(sum == 0.3);
IO.println(Math.abs(sum - 0.3) < 1e-9);

double total = 0;
for (int i = 0; i < 10; i++) {
    total += 0.1; // ten coins of 10 cents, as doubles
}
IO.println("ten times 0.1: " + total);
IO.println("in cents:      " + (10 * 10) + " (a long has no rounding)");
IO.println("BigDecimal:    " + new BigDecimal("0.1").multiply(BigDecimal.TEN));
```
