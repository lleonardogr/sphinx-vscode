## In short

A **logic gate** takes bits in and gives one bit out. **AND** is 1 only when both inputs are 1, **OR** when at least one is, **XOR** when they differ, and **NOT** flips its input. In Java these are `&&`, `||`, `^` and `!` for booleans. **De Morgan's laws** flip a condition: `!(a && b)` is the same as `!a || !b`.

**The truth tables, printed by Java:**

```java
IO.println("a b | AND OR XOR");
for (int a = 0; a <= 1; a++) {
    for (int b = 0; b <= 1; b++) {
        IO.println(a + " " + b + " |  " + (a & b) + "   " + (a | b) + "   " + (a ^ b));
    }
}
```

Gates also do arithmetic. Adding two bits gives a **sum** bit, which is exactly XOR, and a **carry** bit, which is exactly AND. Chain these adders, passing each carry to the next column, and you can add whole numbers: that's how the processor adds.

![A half adder: XOR gives the sum, AND gives the carry](half-adder.svg)

Java can apply a gate to all 32 bits of an `int` at once: `&`, `|`, `^` and `~`. Shifting moves bits: `x << 1` doubles (dropping bits that overflow), and `x >> 1` halves, rounding down: `-3 >> 1` is −2. A **mask** is a number with only the bits you care about turned on; `1 << k` selects bit k:

| Task | Code |
|------|------|
| Is bit k on? | `(n >> k) & 1` |
| Turn it on | `n \| (1 << k)` |
| Turn it off | `n & ~(1 << k)` |
| Flip it | `n ^ (1 << k)` |

**Masks in action.** Change `k` and run again.

```java
int n = 0b1011_0010;
int k = 3;
IO.println("n:          " + Integer.toBinaryString(n));
IO.println("bit k on?   " + ((n >> k) & 1));
IO.println("turn on:    " + Integer.toBinaryString(n | (1 << k)));
IO.println("turn off:   " + Integer.toBinaryString(n & ~(1 << k)));
IO.println("flip:       " + Integer.toBinaryString(n ^ (1 << k)));
IO.println("n << 1:     " + Integer.toBinaryString(n << 1) + " = " + (n << 1));
```

File permissions, colors and network flags all pack several yes/no values into one number this way.
