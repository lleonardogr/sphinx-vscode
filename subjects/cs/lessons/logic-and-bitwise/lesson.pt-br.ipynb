## Em resumo

Uma **porta lógica** recebe bits e devolve um bit. **AND** é 1 só quando as duas entradas são 1, **OR** quando pelo menos uma é, **XOR** quando elas são diferentes, e **NOT** inverte a entrada. No Java elas são `&&`, `||`, `^` e `!` para booleanos. As **leis de De Morgan** invertem uma condição: `!(a && b)` é o mesmo que `!a || !b`.

**As tabelas-verdade, impressas pelo Java:**

````java
IO.println("a b | AND OR XOR");
for (int a = 0; a <= 1; a++) {
    for (int b = 0; b <= 1; b++) {
        IO.println(a + " " + b + " |  " + (a & b) + "   " + (a | b) + "   " + (a ^ b));
    }
}
````

As portas também fazem contas. Somar dois bits dá um bit de **soma**, que é exatamente o XOR, e um bit de **vai um**, que é exatamente o AND. Encadeie esses somadores, passando cada vai-um para a próxima coluna, e você soma números inteiros: é assim que o processador soma.

O Java aplica uma porta aos 32 bits de um `int` de uma vez: `&`, `|`, `^` e `~`. Deslocar move os bits: `x << 1` dobra (descartando os bits que estouram), e `x >> 1` divide ao meio, arredondando para baixo: `-3 >> 1` dá −2. Uma **máscara** é um número só com os bits que interessam ligados; `1 << k` seleciona o bit k:

| Tarefa | Código |
|--------|--------|
| O bit k está ligado? | `(n >> k) & 1` |
| Ligar | `n \| (1 << k)` |
| Desligar | `n & ~(1 << k)` |
| Inverter | `n ^ (1 << k)` |

#### 🎞️ Uma máscara no bit 4

![A máscara 1 << 4 seleciona o bit 4 de 10110010: o AND com ela testa o bit, o AND com o inverso o desliga, dando 10100010](mask-steps.pt-br.svg)

**Máscaras na prática.** Mude `k` e rode de novo.

````java
int n = 0b1011_0010;
int k = 3;
IO.println("n:           " + Integer.toBinaryString(n));
IO.println("bit k ligado? " + ((n >> k) & 1));
IO.println("ligar:       " + Integer.toBinaryString(n | (1 << k)));
IO.println("desligar:    " + Integer.toBinaryString(n & ~(1 << k)));
IO.println("inverter:    " + Integer.toBinaryString(n ^ (1 << k)));
IO.println("n << 1:      " + Integer.toBinaryString(n << 1) + " = " + (n << 1));
````
