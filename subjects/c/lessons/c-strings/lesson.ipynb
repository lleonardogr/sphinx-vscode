## 1. A string is a char array ending in '\0'

C has no `String` type. A string is a `char` array, and the text ends at the special character `'\0'` (the byte 0). `char s[] = "hi";` takes **3** bytes: `'h'`, `'i'` and `'\0'`. Every string function finds the end by walking until the `'\0'`, so it must always be there. The slides show the bytes.

#### 🎞️ A C string, byte by byte

![A C string is a char array ending in '\0': strlen counts up to it, and strcat writes over it and adds a new one at the end](string-bytes-steps.svg)

````c
char s[] = "hi";
printf("sizeof: %zu, strlen: %zu\n", sizeof s, strlen(s));
for (size_t i = 0; i < sizeof s; i++) {
    printf("s[%zu] = '%c' (code %d)\n", i, s[i] ? s[i] : '0', s[i]);
}
printf("%s\n", s);
````

## 2. Room for the '\0'

An array for text needs **one more slot than the longest text**: `char name[32]` fits 31 characters plus `'\0'`. Writing past the end is undefined behavior, as with any array. `snprintf(buffer, sizeof buffer, format, …)` builds text like `printf` but never writes more than the buffer holds, always adding the `'\0'`. It returns the length the whole text would need, so you can tell when it was cut.

````c
char label[16];
snprintf(label, sizeof label, "Room %d-%c", 12, 'B');
printf("[%s] uses %zu of %zu bytes\n", label, strlen(label) + 1, sizeof label);

char small[8];
char sentence[32];
strcpy(sentence, "a long sentence");     // imagine it was typed
int needed = snprintf(small, sizeof small, "%s", sentence);
printf("[%s]: cut to fit; the whole text needed %d chars\n", small, needed);
````

## 3. The tools in <string.h>

| Function | Does |
|---|---|
| `strlen(s)` | the length, not counting `'\0'` |
| `strcmp(a, b)` | 0 when equal; negative or positive for dictionary order |
| `strcpy(dest, src)` | copies src into dest (dest must have room) |
| `strcat(dest, src)` | adds src to the end of dest (room again) |
| `strchr(s, c)` | a pointer to the first `c` in s, or `NULL` |

**Never compare strings with `==`:** it compares the arrays' addresses, not the letters.

````c
char a[20] = "pear";
char b[20] = "pear";
printf("strcmp(a, b) == 0? %d\n", strcmp(a, b) == 0);
printf("strcmp(\"apple\", \"pear\") < 0? %d\n", strcmp("apple", "pear") < 0);
strcat(a, "s");
strcpy(b, "fig");
printf("%s %s (%zu letters)\n", a, b, strlen(a));
char *dash = strchr("2024-11-26", '-');
printf("after the first dash: %s\n", dash + 1);
````

🧪 **Try it:** make `a` only 5 bytes (`char a[5]`) and run: `strcat` writes past the end, and the memory check reports it.

## 4. Changing characters

A `char` array can be changed character by character. Loop until the `'\0'` (no need for `strlen`), and change `s[i]`. Careful with the two ways to write text: `char s[] = "hi";` is your own array, which you can change; `const char *p = "hi";` points to a **literal**, stored in read-only memory: changing it is undefined behavior.

````c
char word[] = "banana";
for (int i = 0; word[i] != '\0'; i++) {
    if (word[i] == 'a') {
        word[i] = 'o';
    }
}
printf("%s\n", word);
const char *fixed = "literal text";
printf("%s\n", fixed);
````

🧪 **Try it:** try `fixed[0] = 'T';` and read the error: `const` protects the literal.

## 5. Reading a line safely

In the challenges, read a line with `fgets(buffer, sizeof buffer, stdin)`: it stops at the newline or when the buffer is full, so it never writes past the end. It **keeps** the `'\n'`, which you usually remove with `buffer[strcspn(buffer, "\n")] = '\0';` (`strcspn` gives the index of the first `'\n'`, or the length when there is none). Below, the "typed" line is written directly.

````c
char line[32] = "Ada Lovelace\n";       // what fgets would leave
printf("before: [%s]\n", line);
line[strcspn(line, "\n")] = '\0';
printf("after:  [%s]\n", line);
````

🧪 **Try it:** remove the `\n` from the initial text and run: `strcspn` gives the length, so nothing breaks.

## 6. Numbers to text and back

`strtol(text, &end, 10)` reads a whole number from text and tells you where it stopped, so you can check that the whole text was a number (`atoi` is simpler but can't report errors). In the other direction, `snprintf` writes a number into a buffer.

````c
const char *inputs[] = {"42", "12abc", "abc"};
for (int i = 0; i < 3; i++) {
    char *end;
    long value = strtol(inputs[i], &end, 10);
    if (end == inputs[i]) {
        printf("%-6s: not a number\n", inputs[i]);
    } else if (*end != '\0') {
        printf("%-6s: %ld, then extra text \"%s\"\n", inputs[i], value, end);
    } else {
        printf("%-6s: %ld\n", inputs[i], value);
    }
}
char buffer[16];
snprintf(buffer, sizeof buffer, "%d", 2026);
printf("%s has %zu digits\n", buffer, strlen(buffer));
````

🧪 **Try it:** add `"-7"` and `" 8"` (with a space) to the inputs. Then print `atoi("12abc")` and `atoi("abc")`: can you tell an error from a zero?

## In short

C has no `String` type: a string is a `char` array whose text ends with the character `'\0'`. Functions find the end by looking for it, so the array needs one extra slot. `<string.h>` has the tools: `strlen`, `strcmp`, `strcpy`, `strchr`.

```c
#include <stdio.h>
#include <string.h>

int main(void) {
    char name[32];                              // up to 31 characters + '\0'
    if (fgets(name, sizeof name, stdin) == NULL) {
        return 1;
    }
    name[strcspn(name, "\n")] = '\0';           // fgets keeps the newline
    if (strcmp(name, "Ada") == 0) {             // == would compare addresses
        printf("Hello again, %s!\n", name);
    }
    printf("%zu letters\n", strlen(name));
    return 0;
}
```

`fgets` stops at the newline or when the array is full, so it never writes past the end. `snprintf(buffer, sizeof buffer, …)` builds text with the same guarantee.

**Watch out:** a string with no `'\0'` keeps going into whatever memory follows. Leave room for it: `"Ada"` needs 4 chars.
