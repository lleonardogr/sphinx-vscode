## In short

C has no `String` type: a string is a `char` array whose text ends with the character `'\0'`. Functions find the end by looking for it, so the array needs one extra slot. `<string.h>` has the tools: `strlen`, `strcmp`, `strcpy`, `strchr`.

#### 🎞️ A C string, byte by byte

![A C string is a char array ending in '\0': strlen counts up to it, and strcat writes over it and adds a new one at the end](string-bytes-steps.svg)

```c
#include <stdio.h>
#include <string.h>

int main(void) {
    char name[32];                              // up to 31 characters + '\0'
    if (fgets(name, sizeof name, stdin) == NULL) {
        return 1;
    }
    name[strcspn(name, "\n")] = '\0';           // fgets keeps the newline
    if (strcmp(name, "Ada") == 0) {             // == would compare addresses
        printf("Hello again, %s!\n", name);
    }
    printf("%zu letters\n", strlen(name));
    return 0;
}
```

`fgets` stops at the newline or when the array is full, so it never writes past the end. `snprintf(buffer, sizeof buffer, …)` builds text with the same guarantee.

**Strings are char arrays:** the length doesn't count the `'\0'`, but the array needs room for it.

````c
char word[16] = "hello";
printf("strlen %zu, sizeof %zu\n", strlen(word), sizeof word);
printf("the char after the text: %d\n", word[5]);
strcat(word, " you");
printf("%s (%zu)\n", word, strlen(word));
printf("%d %d\n", strcmp("apple", "banana") < 0, strcmp("abc", "abc") == 0);
for (int i = 0; word[i] != '\0'; i++) {
    printf("%c", word[i] == ' ' ? '_' : word[i]);
}
printf("\n");
````

void to_upper_str(char *s) {
    for (int i = 0; s[i] != '\0'; i++) {
        if (s[i] >= 'a' && s[i] <= 'z') {
            s[i] = (char) (s[i] - 'a' + 'A');
        }
    }
}
